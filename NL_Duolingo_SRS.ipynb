```python
# ============================================
# TRACES DATASET - INITIAL CLEANING AUDIT
# ============================================

print("=" * 60)
print("TRACES DATASET - BASIC INFORMATION")
print("=" * 60)

print(f"Rows    : {traces_df.shape[0]:,}")
print(f"Columns : {traces_df.shape[1]}")

print("\nColumn Names:")
print(traces_df.columns.tolist())

print("\nData Types:")
print(traces_df.dtypes)

print("\nMissing Values:")
print(traces_df.isnull().sum())

print("\nDuplicate Rows:")
print(traces_df.duplicated().sum())
```

```python
# ============================================
# STEP 1.2 - NUMERICAL VALIDATION
# ============================================

print("=" * 60)
print("NUMERICAL VALIDATION")
print("=" * 60)

# --------------------------------------------
# 1. Check negative lag_days
# --------------------------------------------

negative_lag = (traces_df["lag_days"] < 0).sum()

print(f"\n1. Negative lag_days      : {negative_lag:,}")


# --------------------------------------------
# 2. Check negative history counts
# --------------------------------------------

negative_history_seen = (traces_df["history_seen"] < 0).sum()
negative_history_correct = (traces_df["history_correct"] < 0).sum()

print(f"2. Negative history_seen  : {negative_history_seen:,}")
print(f"3. Negative history_correct: {negative_history_correct:,}")


# --------------------------------------------
# 3. Check history_correct > history_seen
# --------------------------------------------

invalid_history = (
    traces_df["history_correct"] > traces_df["history_seen"]
).sum()

print(f"4. history_correct > history_seen : {invalid_history:,}")


# --------------------------------------------
# 4. Check negative session counts
# --------------------------------------------

negative_session_seen = (traces_df["session_seen"] < 0).sum()
negative_session_correct = (traces_df["session_correct"] < 0).sum()

print(f"5. Negative session_seen  : {negative_session_seen:,}")
print(f"6. Negative session_correct: {negative_session_correct:,}")


# --------------------------------------------
# 5. Check session_correct > session_seen
# --------------------------------------------

invalid_session = (
    traces_df["session_correct"] > traces_df["session_seen"]
).sum()

print(f"7. session_correct > session_seen : {invalid_session:,}")


# --------------------------------------------
# 6. Check p_recall outside [0, 1]
# --------------------------------------------

invalid_p_recall = (
    (traces_df["p_recall"] < 0) |
    (traces_df["p_recall"] > 1)
).sum()

print(f"8. p_recall outside [0, 1] : {invalid_p_recall:,}")


# --------------------------------------------
# 7. Display numerical summary
# --------------------------------------------

print("\n" + "=" * 60)
print("NUMERICAL SUMMARY")
print("=" * 60)

print(
    traces_df[
        [
            "lag_days",
            "history_seen",
            "history_correct",
            "session_seen",
            "session_correct",
            "p_recall"
        ]
    ].describe().round(3)
)
```

```python
# ============================================
# STEP 1.3 - CATEGORICAL DATA VALIDATION
# ============================================

print("=" * 60)
print("CATEGORICAL DATA VALIDATION")
print("=" * 60)

# --------------------------------------------
# Unique value counts
# --------------------------------------------

categorical_columns = [
    "user_id",
    "ui_language",
    "learning_language",
    "surface_form",
    "lemma",
    "pos",
    "grammar_tags",
    "lexeme_id"
]

print("\nUnique Value Counts:")
print("-" * 60)

for col in categorical_columns:
    print(f"{col:<20}: {traces_df[col].nunique(dropna=True):,}")


# --------------------------------------------
# Check blank / empty strings
# --------------------------------------------

print("\nBlank / Empty String Counts:")
print("-" * 60)

for col in categorical_columns:
    blank_count = (
        traces_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(f"{col:<20}: {blank_count:,}")


# --------------------------------------------
# UI languages
# --------------------------------------------

print("\nUI Languages:")
print("-" * 60)
print(traces_df["ui_language"].value_counts(dropna=False))


# --------------------------------------------
# Learning languages
# --------------------------------------------

print("\nLearning Languages:")
print("-" * 60)
print(traces_df["learning_language"].value_counts(dropna=False))


# --------------------------------------------
# POS categories
# --------------------------------------------

print("\nPOS Categories:")
print("-" * 60)
print(traces_df["pos"].value_counts(dropna=False))
```

```python
# ============================================
# STEP 1.4 - CONVERT PRACTICE_TIME TO DATETIME
# ============================================

traces_df["practice_time"] = pd.to_datetime(
    traces_df["practice_time"],
    errors="coerce"
)

print("=" * 60)
print("PRACTICE_TIME DATETIME CONVERSION")
print("=" * 60)

print("\nData Type:")
print(traces_df["practice_time"].dtype)

print("\nMissing Values After Conversion:")
print(traces_df["practice_time"].isna().sum())

print("\nDate Range:")
print("Minimum:", traces_df["practice_time"].min())
print("Maximum:", traces_df["practice_time"].max())

print("\nSample Values:")
print(traces_df["practice_time"].head())
```

```python
# ============================================
# STEP 1.5 - P_RECALL CONSISTENCY CHECK
# ============================================

# Calculate observed session recall
traces_df["calculated_recall"] = (
    traces_df["session_correct"] / traces_df["session_seen"]
)

# Compare calculated recall with existing p_recall
recall_difference = (
    traces_df["p_recall"] - traces_df["calculated_recall"]
).abs()

print("=" * 60)
print("P_RECALL CONSISTENCY CHECK")
print("=" * 60)

print("\nMaximum absolute difference:")
print(recall_difference.max())

print("\nNumber of rows with difference > 0.000001:")
print((recall_difference > 0.000001).sum())

print("\nNumber of rows with difference > 0.001:")
print((recall_difference > 0.001).sum())

print("\nSample comparison:")
print(
    traces_df[
        [
            "session_seen",
            "session_correct",
            "p_recall",
            "calculated_recall"
        ]
    ].head(10)
)
```

```python
# Remove temporary validation column
traces_df.drop(columns=["calculated_recall"], inplace=True)

print("Temporary validation column removed.")
print(f"Current shape: {traces_df.shape}")
```

```python
# ============================================
# STEP 1.6 - FINAL CLEANING VALIDATION
# ============================================

print("=" * 60)
print("FINAL CLEANING VALIDATION - TRACES_DF")
print("=" * 60)

print(f"\nRows    : {traces_df.shape[0]:,}")
print(f"Columns : {traces_df.shape[1]}")

print("\nData Types:")
print(traces_df.dtypes)

print("\nMissing Values:")
print(traces_df.isnull().sum())

print(f"\nDuplicate Rows: {traces_df.duplicated().sum()}")

print("\nDate Range:")
print(f"Start: {traces_df['practice_time'].min()}")
print(f"End  : {traces_df['practice_time'].max()}")
```

```python
# ============================================
# STEP 1.7 - SAVE CLEANED TRACES DATASET
# ============================================

output_file = "/content/traces_cleaned.csv"

traces_df.to_csv(output_file, index=False)

print("=" * 60)
print("CLEANED DATASET SAVED SUCCESSFULLY")
print("=" * 60)

print(f"\nFile: {output_file}")
print(f"Rows: {traces_df.shape[0]:,}")
print(f"Columns: {traces_df.shape[1]}")
```

```python
# ============================================
# WORD DIFFICULTY - INITIAL CLEANING AUDIT
# ============================================

print("=" * 60)
print("WORD DIFFICULTY - BASIC INFORMATION")
print("=" * 60)

print(f"Rows    : {word_difficulty_df.shape[0]:,}")
print(f"Columns : {word_difficulty_df.shape[1]}")

print("\nColumn Names:")
print(word_difficulty_df.columns.tolist())

print("\nData Types:")
print(word_difficulty_df.dtypes)

print("\nMissing Values:")
print(word_difficulty_df.isnull().sum())

print("\nDuplicate Rows:")
print(word_difficulty_df.duplicated().sum())

print("\nFirst 5 Rows:")
print(word_difficulty_df.head())
```

```python
# ============================================
# STEP 2.2 - NUMERICAL VALIDATION
# ============================================

print("=" * 60)
print("WORD DIFFICULTY - NUMERICAL VALIDATION")
print("=" * 60)


# --------------------------------------------
# 1. Check n_traces
# --------------------------------------------

invalid_n_traces = (
    word_difficulty_df["n_traces"] <= 0
).sum()

print(f"\n1. n_traces <= 0 : {invalid_n_traces:,}")


# --------------------------------------------
# 2. Check avg_session_recall
# --------------------------------------------

invalid_session_recall = (
    (word_difficulty_df["avg_session_recall"] < 0) |
    (word_difficulty_df["avg_session_recall"] > 1)
).sum()

print(
    f"2. avg_session_recall outside [0,1] : "
    f"{invalid_session_recall:,}"
)


# --------------------------------------------
# 3. Check item_recall_rate
# --------------------------------------------

invalid_item_recall = (
    (word_difficulty_df["item_recall_rate"] < 0) |
    (word_difficulty_df["item_recall_rate"] > 1)
).sum()

print(
    f"3. item_recall_rate outside [0,1] : "
    f"{invalid_item_recall:,}"
)


# --------------------------------------------
# 4. Check avg_prior_exposures
# --------------------------------------------

invalid_exposures = (
    word_difficulty_df["avg_prior_exposures"] < 0
).sum()

print(
    f"4. avg_prior_exposures < 0 : "
    f"{invalid_exposures:,}"
)


# --------------------------------------------
# 5. Check difficulty rank
# --------------------------------------------

invalid_rank = (
    word_difficulty_df["difficulty_rank_in_language"] < 1
).sum()

print(
    f"5. difficulty_rank_in_language < 1 : "
    f"{invalid_rank:,}"
)


# --------------------------------------------
# 6. Numerical summary
# --------------------------------------------

print("\n" + "=" * 60)
print("NUMERICAL SUMMARY")
print("=" * 60)

print(
    word_difficulty_df[
        [
            "n_traces",
            "avg_session_recall",
            "item_recall_rate",
            "avg_prior_exposures",
            "difficulty_rank_in_language"
        ]
    ].describe().round(3)
)
```

```python
# ============================================
# STEP 2.3 - LANGUAGE & RANK CONSISTENCY
# ============================================

print("=" * 60)
print("LANGUAGE & DIFFICULTY RANK VALIDATION")
print("=" * 60)


# --------------------------------------------
# 1. Learning language distribution
# --------------------------------------------

print("\nLearning Language Distribution:")
print("-" * 60)

language_counts = (
    word_difficulty_df["learning_language"]
    .value_counts()
    .sort_index()
)

print(language_counts)


# --------------------------------------------
# 2. Language code → language name mapping
# --------------------------------------------

print("\nLanguage Code → Language Name Mapping:")
print("-" * 60)

language_mapping = (
    word_difficulty_df[
        ["learning_language", "language_name"]
    ]
    .drop_duplicates()
    .sort_values(["learning_language", "language_name"])
)

print(language_mapping.to_string(index=False))


# --------------------------------------------
# 3. Check whether one language code maps
#    to multiple language names
# --------------------------------------------

mapping_check = (
    word_difficulty_df
    .groupby("learning_language")["language_name"]
    .nunique()
)

print("\nLanguage Codes Mapping to Multiple Names:")
print("-" * 60)

print(mapping_check[mapping_check > 1])


# --------------------------------------------
# 4. Difficulty rank summary by language
# --------------------------------------------

print("\nDifficulty Rank by Learning Language:")
print("-" * 60)

rank_summary = (
    word_difficulty_df
    .groupby("learning_language")["difficulty_rank_in_language"]
    .agg(
        word_count="count",
        min_rank="min",
        max_rank="max",
        unique_ranks="nunique"
    )
    .sort_index()
)

print(rank_summary)


# --------------------------------------------
# 5. Check duplicate ranks within language
# --------------------------------------------

duplicate_ranks = (
    word_difficulty_df
    .groupby(
        ["learning_language", "difficulty_rank_in_language"]
    )
    .size()
    .reset_index(name="count")
)

duplicate_ranks = duplicate_ranks[
    duplicate_ranks["count"] > 1
]

print("\nDuplicate Difficulty Ranks Within Language:")
print("-" * 60)

print(f"Number of duplicated rank groups: {len(duplicate_ranks):,}")

if len(duplicate_ranks) > 0:
    print("\nSample:")
    print(duplicate_ranks.head(10).to_string(index=False))
```

```python
# ============================================
# STEP 2.4 - DIFFICULTY RANK STRUCTURE
# ============================================

print("=" * 60)
print("DIFFICULTY RANK STRUCTURE")
print("=" * 60)

for language, group in word_difficulty_df.groupby("learning_language"):

    ranks = sorted(
        group["difficulty_rank_in_language"].unique()
    )

    expected_ranks = set(
        range(1, max(ranks) + 1)
    )

    actual_ranks = set(ranks)

    missing_ranks = sorted(
        expected_ranks - actual_ranks
    )

    print(f"\nLanguage: {language}")
    print(f"Number of words       : {len(group):,}")
    print(f"Minimum rank          : {min(ranks)}")
    print(f"Maximum rank          : {max(ranks)}")
    print(f"Unique ranks          : {len(ranks):,}")
    print(f"Missing rank numbers  : {len(missing_ranks):,}")

    if len(missing_ranks) > 0:
        print(f"Sample missing ranks  : {missing_ranks[:10]}")
    else:
        print("Missing rank numbers  : None")
```

```python
# ============================================
# STEP 2.5 - CATEGORICAL VALIDATION
# ============================================

print("=" * 60)
print("WORD DIFFICULTY - CATEGORICAL VALIDATION")
print("=" * 60)


categorical_columns = [
    "learning_language",
    "language_name",
    "surface_form",
    "lemma",
    "pos",
    "grammar_tags",
    "lexeme_id"
]


# --------------------------------------------
# 1. Unique values
# --------------------------------------------

print("\nUnique Value Counts:")
print("-" * 60)

for col in categorical_columns:
    print(
        f"{col:<25}: "
        f"{word_difficulty_df[col].nunique(dropna=True):,}"
    )


# --------------------------------------------
# 2. Blank / empty strings
# --------------------------------------------

print("\nBlank / Empty String Counts:")
print("-" * 60)

for col in categorical_columns:
    blank_count = (
        word_difficulty_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(f"{col:<25}: {blank_count:,}")


# --------------------------------------------
# 3. Duplicate lexeme IDs
# --------------------------------------------

duplicate_lexeme_ids = (
    word_difficulty_df["lexeme_id"]
    .duplicated()
    .sum()
)

print("\nDuplicate Lexeme IDs:")
print("-" * 60)
print(f"Duplicate lexeme_id rows: {duplicate_lexeme_ids:,}")


# --------------------------------------------
# 4. POS distribution
# --------------------------------------------

print("\nPOS Distribution:")
print("-" * 60)

print(
    word_difficulty_df["pos"]
    .value_counts(dropna=False)
)


# --------------------------------------------
# 5. Grammar tag missingness
# --------------------------------------------

print("\nGrammar Tags:")
print("-" * 60)

print(
    f"Missing grammar_tags: "
    f"{word_difficulty_df['grammar_tags'].isna().sum():,}"
)

print(
    f"Non-missing grammar_tags: "
    f"{word_difficulty_df['grammar_tags'].notna().sum():,}"
)


# --------------------------------------------
# 6. Language distribution
# --------------------------------------------

print("\nLanguage Distribution:")
print("-" * 60)

print(
    word_difficulty_df[
        ["learning_language", "language_name"]
    ]
    .drop_duplicates()
    .sort_values("learning_language")
    .to_string(index=False)
)
```

```python
# ============================================
# STEP 2.6 - FINAL WORD-LEVEL SANITY CHECK
# ============================================

print("=" * 60)
print("WORD DIFFICULTY - FINAL SANITY CHECK")
print("=" * 60)


# --------------------------------------------
# 1. Summary by learning language
# --------------------------------------------

print("\nSummary by Learning Language:")
print("-" * 60)

language_summary = (
    word_difficulty_df
    .groupby("learning_language")
    .agg(
        words=("lexeme_id", "count"),
        total_traces=("n_traces", "sum"),
        avg_session_recall=("avg_session_recall", "mean"),
        avg_item_recall=("item_recall_rate", "mean"),
        avg_prior_exposures=("avg_prior_exposures", "mean"),
        min_recall=("item_recall_rate", "min"),
        max_recall=("item_recall_rate", "max")
    )
    .round(4)
)

print(language_summary)


# --------------------------------------------
# 2. Check recall relationship
# --------------------------------------------

recall_difference = (
    word_difficulty_df["item_recall_rate"]
    - word_difficulty_df["avg_session_recall"]
)

print("\nItem Recall Rate - Average Session Recall:")
print("-" * 60)

print(
    recall_difference.describe().round(4)
)


# --------------------------------------------
# 3. Largest positive differences
# --------------------------------------------

print("\nLargest Positive Differences:")
print("-" * 60)

positive_examples = (
    word_difficulty_df
    .assign(
        recall_difference=recall_difference
    )
    .sort_values(
        "recall_difference",
        ascending=False
    )
    [
        [
            "learning_language",
            "surface_form",
            "lemma",
            "n_traces",
            "avg_session_recall",
            "item_recall_rate",
            "recall_difference"
        ]
    ]
    .head(5)
)

print(positive_examples.to_string(index=False))


# --------------------------------------------
# 4. Largest negative differences
# --------------------------------------------

print("\nLargest Negative Differences:")
print("-" * 60)

negative_examples = (
    word_difficulty_df
    .assign(
        recall_difference=recall_difference
    )
    .sort_values(
        "recall_difference",
        ascending=True
    )
    [
        [
            "learning_language",
            "surface_form",
            "lemma",
            "n_traces",
            "avg_session_recall",
            "item_recall_rate",
            "recall_difference"
        ]
    ]
    .head(5)
)

print(negative_examples.to_string(index=False))
```

```python
# ============================================
# STEP 2.7 - FINAL CLEANING VALIDATION
# ============================================

print("=" * 60)
print("FINAL CLEANING VALIDATION - WORD DIFFICULTY")
print("=" * 60)

print(f"\nRows    : {word_difficulty_df.shape[0]:,}")
print(f"Columns : {word_difficulty_df.shape[1]}")

print("\nData Types:")
print(word_difficulty_df.dtypes)

print("\nMissing Values:")
print(word_difficulty_df.isnull().sum())

print(f"\nDuplicate Rows: {word_difficulty_df.duplicated().sum()}")

print(
    f"\nUnique Lexeme IDs: "
    f"{word_difficulty_df['lexeme_id'].nunique():,}"
)

print(
    f"\nLearning Languages: "
    f"{word_difficulty_df['learning_language'].nunique():,}"
)
```

```python
# ============================================================
# Create lag bins in the raw traces dataset
# ============================================================

import numpy as np

# Convert lag from days to hours
lag_hours = traces_cleaned['lag_days'] * 24

# Define lag boundaries in hours
lag_bins = [
    -np.inf,
    1,
    6,
    24,
    72,
    168,
    336,
    672,
    2160,
    np.inf
]

lag_labels = [
    '<1 hour',
    '1-6 hours',
    '6-24 hours',
    '1-3 days',
    '3-7 days',
    '1-2 weeks',
    '2-4 weeks',
    '1-3 months',
    '3+ months'
]

traces_cleaned['lag_bin'] = pd.cut(
    lag_hours,
    bins=lag_bins,
    labels=lag_labels,
    right=False,
    ordered=True
)

# Check the distribution
traces_cleaned['lag_bin'].value_counts().sort_index()
```

```python
# ============================================================
# Recall by lag bin using the raw recall attempts
# ============================================================

lag_recall_raw = (
    traces_cleaned
    .groupby('lag_bin', observed=False)
    .agg(
        n_traces=('user_id', 'size'),
        total_attempts=('session_seen', 'sum'),
        total_correct=('session_correct', 'sum')
    )
    .reset_index()
)

# Calculate Word Recall Rate from raw attempts
lag_recall_raw['item_recall_rate'] = (
    lag_recall_raw['total_correct']
    / lag_recall_raw['total_attempts']
)

# Calculate incorrect attempts
lag_recall_raw['total_incorrect'] = (
    lag_recall_raw['total_attempts']
    - lag_recall_raw['total_correct']
)

lag_recall_raw['incorrect_rate'] = (
    lag_recall_raw['total_incorrect']
    / lag_recall_raw['total_attempts']
)

lag_recall_raw
```

```python
# ============================================================
# H1 — SPEARMAN CORRELATION
# Does recall decrease as lag_days increases?
# ============================================================

from scipy.stats import spearmanr

# Remove rows with missing values in the variables being tested
h1_data = traces_cleaned[
    ['lag_days', 'calculated_recall']
].dropna()

# Spearman correlation
rho, p_value = spearmanr(
    h1_data['lag_days'],
    h1_data['calculated_recall']
)

print("H1 — TIME SINCE LAST PRACTICE → RECALL")
print("=" * 60)

print(f"Observations tested : {len(h1_data):,}")
print(f"Spearman's rho      : {rho:.6f}")
print(f"P-value             : {p_value:.10f}")

print("\nHypotheses")
print("-" * 60)
print("H0: No negative relationship between lag_days and recall")
print("H1: Recall decreases as lag_days increases")

print("\nDecision rule")
print("-" * 60)
print("Significance level (α): 0.05")

if p_value < 0.05:
    print("Result: Statistically significant relationship")
else:
    print("Result: Not statistically significant")
```

```python
# ============================================================
# H1 — STEP 2
# BINOMIAL LOGISTIC REGRESSION
# Effect of time since last practice on recall
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm

# ------------------------------------------------------------
# 1. Prepare data
# ------------------------------------------------------------

h1_model = traces_cleaned[
    ['lag_days', 'session_seen', 'session_correct']
].copy()

# Remove missing values
h1_model = h1_model.dropna()

# Keep valid observations only
h1_model = h1_model[
    (h1_model['session_seen'] > 0) &
    (h1_model['session_correct'] >= 0) &
    (h1_model['session_correct'] <= h1_model['session_seen']) &
    (h1_model['lag_days'] >= 0)
].copy()

# ------------------------------------------------------------
# 2. Log-transform lag
# ------------------------------------------------------------
# log1p handles lag = 0 safely

h1_model['log_lag'] = np.log1p(h1_model['lag_days'])

# ------------------------------------------------------------
# 3. Binomial logistic regression
# ------------------------------------------------------------

X = sm.add_constant(h1_model[['log_lag']])

y_success = h1_model['session_correct']
y_failure = h1_model['session_seen'] - h1_model['session_correct']

model = sm.GLM(
    np.column_stack([y_success, y_failure]),
    X,
    family=sm.families.Binomial()
)

result = model.fit()

# ------------------------------------------------------------
# 4. Extract results
# ------------------------------------------------------------

coef = result.params['log_lag']
p_value = result.pvalues['log_lag']

odds_ratio = np.exp(coef)

ci_lower = np.exp(result.conf_int().loc['log_lag', 0])
ci_upper = np.exp(result.conf_int().loc['log_lag', 1])

# ------------------------------------------------------------
# 5. Display
# ------------------------------------------------------------

print("H1 — EFFECT OF TIME SINCE LAST PRACTICE ON RECALL")
print("=" * 65)

print(f"Observations used       : {len(h1_model):,}")
print(f"Coefficient (log lag)   : {coef:.6f}")
print(f"Odds ratio              : {odds_ratio:.6f}")
print(f"95% CI for odds ratio   : {ci_lower:.6f} to {ci_upper:.6f}")
print(f"P-value                 : {p_value:.10f}")

print("\nHypotheses")
print("-" * 65)
print("H0: Time since last practice has no effect on recall")
print("H1: Longer time since last practice reduces recall")

print("\nDecision")
print("-" * 65)

if p_value < 0.05 and coef < 0:
    print("Reject H0")
    print("Evidence supports a negative effect of lag on recall.")
elif p_value < 0.05 and coef > 0:
    print("Reject H0")
    print("A significant relationship exists, but the direction is positive.")
else:
    print("Fail to reject H0")
    print("No statistically significant effect detected.")

print("\nModel interpretation")
print("-" * 65)

if odds_ratio < 1:
    print(
        f"For a one-unit increase in log(1 + lag_days), "
        f"the odds of a correct recall are multiplied by "
        f"{odds_ratio:.4f}."
    )
else:
    print(
        f"For a one-unit increase in log(1 + lag_days), "
        f"the odds of a correct recall are multiplied by "
        f"{odds_ratio:.4f}."
    )
```

```python
# ============================================================
# H1 — STEP 3
# PRACTICAL RECALL DECAY BY TIME SINCE LAST PRACTICE
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Aggregate recall by lag bin
# ------------------------------------------------------------

h1_lag_summary = (
    traces_cleaned
    .groupby('lag_bin')
    .agg(
        total_attempts=('session_seen', 'sum'),
        total_correct=('session_correct', 'sum'),
        n_traces=('session_seen', 'size')
    )
    .reset_index()
)

# Calculate recall
h1_lag_summary['recall_%'] = (
    h1_lag_summary['total_correct']
    / h1_lag_summary['total_attempts']
    * 100
)

# Calculate incorrect rate
h1_lag_summary['incorrect_%'] = (
    100 - h1_lag_summary['recall_%']
)

# ------------------------------------------------------------
# 2. Put lag bins into logical chronological order
# ------------------------------------------------------------

lag_order = [
    '<1 hour',
    '1-6 hours',
    '6-24 hours',
    '1-3 days',
    '3-7 days',
    '1-2 weeks',
    '2-4 weeks',
    '1-3 months',
    '3+ months'
]

h1_lag_summary['lag_bin'] = pd.Categorical(
    h1_lag_summary['lag_bin'],
    categories=lag_order,
    ordered=True
)

h1_lag_summary = (
    h1_lag_summary
    .sort_values('lag_bin')
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 3. Calculate recall decay from the shortest lag
# ------------------------------------------------------------

baseline_recall = h1_lag_summary.loc[0, 'recall_%']

h1_lag_summary['recall_loss_vs_baseline_pp'] = (
    h1_lag_summary['recall_%'] - baseline_recall
)

h1_lag_summary['recall_loss_abs_pp'] = (
    baseline_recall - h1_lag_summary['recall_%']
)

# ------------------------------------------------------------
# 4. Display
# ------------------------------------------------------------

print("H1 — PRACTICAL RECALL DECAY BY TIME SINCE LAST PRACTICE")
print("=" * 85)

display(
    h1_lag_summary[
        [
            'lag_bin',
            'total_attempts',
            'total_correct',
            'recall_%',
            'incorrect_%',
            'recall_loss_abs_pp'
        ]
    ].round(2)
)

# ------------------------------------------------------------
# 5. Identify key points
# ------------------------------------------------------------

highest_recall = h1_lag_summary.loc[
    h1_lag_summary['recall_%'].idxmax()
]

lowest_recall = h1_lag_summary.loc[
    h1_lag_summary['recall_%'].idxmin()
]

print("\nKEY FINDINGS")
print("=" * 85)

print(
    f"Highest recall : {highest_recall['lag_bin']} "
    f"→ {highest_recall['recall_%']:.2f}%"
)

print(
    f"Lowest recall  : {lowest_recall['lag_bin']} "
    f"→ {lowest_recall['recall_%']:.2f}%"
)

print(
    f"Maximum recall decline from baseline: "
    f"{lowest_recall['recall_loss_abs_pp']:.2f} percentage points"
)
```

```python
# ============================================================
# H1 — STEP 4
# RECALL DECAY BY LAG BIN AND LEARNING LANGUAGE
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Aggregate recall by language × lag bin
# ------------------------------------------------------------

h1_language_lag = (
    traces_cleaned
    .groupby(['learning_language', 'lag_bin'])
    .agg(
        total_attempts=('session_seen', 'sum'),
        total_correct=('session_correct', 'sum'),
        n_traces=('session_seen', 'size')
    )
    .reset_index()
)

# Recall
h1_language_lag['recall_%'] = (
    h1_language_lag['total_correct']
    / h1_language_lag['total_attempts']
    * 100
)

# ------------------------------------------------------------
# 2. Order lag bins chronologically
# ------------------------------------------------------------

lag_order = [
    '<1 hour',
    '1-6 hours',
    '6-24 hours',
    '1-3 days',
    '3-7 days',
    '1-2 weeks',
    '2-4 weeks',
    '1-3 months',
    '3+ months'
]

h1_language_lag['lag_bin'] = pd.Categorical(
    h1_language_lag['lag_bin'],
    categories=lag_order,
    ordered=True
)

h1_language_lag = (
    h1_language_lag
    .sort_values(['learning_language', 'lag_bin'])
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 3. Calculate recall loss from each language's
#    <1 hour baseline
# ------------------------------------------------------------

baseline_by_language = (
    h1_language_lag[
        h1_language_lag['lag_bin'] == '<1 hour'
    ][['learning_language', 'recall_%']]
    .rename(columns={'recall_%': 'baseline_recall_%'})
)

h1_language_lag = h1_language_lag.merge(
    baseline_by_language,
    on='learning_language',
    how='left'
)

h1_language_lag['recall_loss_vs_baseline_pp'] = (
    h1_language_lag['baseline_recall_%']
    - h1_language_lag['recall_%']
)

# ------------------------------------------------------------
# 4. Display detailed table
# ------------------------------------------------------------

print("H1 — RECALL DECAY BY LANGUAGE AND TIME SINCE PRACTICE")
print("=" * 90)

display(
    h1_language_lag[
        [
            'learning_language',
            'lag_bin',
            'total_attempts',
            'recall_%',
            'recall_loss_vs_baseline_pp'
        ]
    ].round(2)
)

# ------------------------------------------------------------
# 5. Summarise overall decay for each language
# ------------------------------------------------------------

language_decay = (
    h1_language_lag
    .groupby('learning_language')
    .agg(
        baseline_recall_pct=('baseline_recall_%', 'first'),
        min_recall_pct=('recall_%', 'min'),
        max_recall_loss_pp=('recall_loss_vs_baseline_pp', 'max'),
        total_attempts=('total_attempts', 'sum')
    )
    .reset_index()
    .sort_values('max_recall_loss_pp', ascending=False)
)

print("\nLANGUAGE-LEVEL DECAY SUMMARY")
print("=" * 90)

display(
    language_decay.round(2)
)
```

```python
# ============================================================
# H1 — FINAL TEST
# DOES LANGUAGE CHANGE THE EFFECT OF TIME SINCE PRACTICE?
# Lag × Learning Language Interaction
# ============================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import chi2

# ------------------------------------------------------------
# 1. Prepare data
# ------------------------------------------------------------

h1_interaction = traces_cleaned.copy()

# Remove rows where required variables are missing
h1_interaction = h1_interaction.dropna(
    subset=[
        'lag_days',
        'learning_language',
        'session_seen',
        'session_correct'
    ]
)

# Keep valid recall observations
h1_interaction = h1_interaction[
    h1_interaction['session_seen'] > 0
]

# Log-transform lag
h1_interaction['log_lag'] = np.log1p(
    h1_interaction['lag_days']
)

# Recall proportion for each trace
h1_interaction['recall_rate'] = (
    h1_interaction['session_correct']
    / h1_interaction['session_seen']
)

# ------------------------------------------------------------
# 2. Model 1: Lag + Language
#    Assumes the lag effect is the same across languages
# ------------------------------------------------------------

model_no_interaction = smf.glm(
    formula="""
        recall_rate ~ log_lag + C(learning_language)
    """,
    data=h1_interaction,
    family=sm.families.Binomial(),
    freq_weights=h1_interaction['session_seen']
).fit()

# ------------------------------------------------------------
# 3. Model 2: Lag × Language
#    Allows lag effect to differ by language
# ------------------------------------------------------------

model_interaction = smf.glm(
    formula="""
        recall_rate ~ log_lag * C(learning_language)
    """,
    data=h1_interaction,
    family=sm.families.Binomial(),
    freq_weights=h1_interaction['session_seen']
).fit()

# ------------------------------------------------------------
# 4. Likelihood Ratio Test
# ------------------------------------------------------------

lr_stat = (
    model_no_interaction.deviance
    - model_interaction.deviance
)

df_diff = (
    model_no_interaction.df_resid
    - model_interaction.df_resid
)

p_value = chi2.sf(
    lr_stat,
    df_diff
)

# ------------------------------------------------------------
# 5. Display model comparison
# ------------------------------------------------------------

print("H1 — LANGUAGE × LAG INTERACTION TEST")
print("=" * 75)

print(f"Observations used        : {len(h1_interaction):,}")
print(f"Model 1 deviance        : {model_no_interaction.deviance:.2f}")
print(f"Model 2 deviance        : {model_interaction.deviance:.2f}")
print(f"Likelihood ratio stat   : {lr_stat:.2f}")
print(f"Degrees of freedom      : {int(df_diff)}")
print(f"P-value                 : {p_value:.10f}")

print("\nHypotheses")
print("-" * 75)
print("H0: The effect of time since practice is the same across languages")
print("H1: The effect of time since practice differs across languages")

print("\nDecision")
print("-" * 75)

if p_value < 0.05:
    print("Reject H0")
    print("Evidence suggests that lag-related recall decay differs by language.")
else:
    print("Fail to reject H0")
    print("No statistically significant evidence that lag decay differs by language.")

# ------------------------------------------------------------
# 6. Show interaction coefficients
# ------------------------------------------------------------

interaction_mask = (
    model_interaction.params.index
    .str.contains(r"log_lag:C\(learning_language\)")
)

interaction_results = pd.DataFrame({
    'coefficient': model_interaction.params[interaction_mask],
    'p_value': model_interaction.pvalues[interaction_mask]
})

print("\nLANGUAGE INTERACTION EFFECTS")
print("=" * 75)

display(
    interaction_results.round(6)
)
```

```python
# ============================================================
# H2 — STEP 3
# MULTIVARIABLE MODEL:
# PRIOR EXPOSURE → RECALL
# CONTROLLING FOR LAG + LANGUAGE
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf

# ------------------------------------------------------------
# 1. Prepare modelling dataset
# ------------------------------------------------------------

h2_model_data = traces_cleaned[
    [
        'history_seen',
        'lag_days',
        'learning_language',
        'session_seen',
        'session_correct'
    ]
].copy()

# Remove invalid observations
h2_model_data = h2_model_data[
    (h2_model_data['session_seen'] > 0) &
    (h2_model_data['session_correct'] >= 0) &
    (h2_model_data['session_correct'] <= h2_model_data['session_seen']) &
    (h2_model_data['history_seen'] >= 1) &
    (h2_model_data['lag_days'] >= 0)
].dropna()

# ------------------------------------------------------------
# 2. Transform skewed variables
# ------------------------------------------------------------

h2_model_data['log_exposure'] = np.log1p(
    h2_model_data['history_seen']
)

h2_model_data['log_lag'] = np.log1p(
    h2_model_data['lag_days']
)

h2_model_data['recall_rate'] = (
    h2_model_data['session_correct']
    / h2_model_data['session_seen']
)

# ------------------------------------------------------------
# 3. Model 1 — linear exposure effect
# ------------------------------------------------------------

model_linear = smf.glm(
    formula="""
        recall_rate
        ~ log_exposure
        + log_lag
        + C(learning_language)
    """,
    data=h2_model_data,
    family=sm.families.Binomial(),
    freq_weights=h2_model_data['session_seen']
).fit()

# ------------------------------------------------------------
# 4. Model 2 — non-linear exposure effect
# ------------------------------------------------------------

h2_model_data['log_exposure_sq'] = (
    h2_model_data['log_exposure'] ** 2
)

model_nonlinear = smf.glm(
    formula="""
        recall_rate
        ~ log_exposure
        + log_exposure_sq
        + log_lag
        + C(learning_language)
    """,
    data=h2_model_data,
    family=sm.families.Binomial(),
    freq_weights=h2_model_data['session_seen']
).fit()

# ------------------------------------------------------------
# 5. Compare models
# ------------------------------------------------------------

lr_stat = model_linear.deviance - model_nonlinear.deviance
df_diff = model_linear.df_resid - model_nonlinear.df_resid

from scipy.stats import chi2

lr_p_value = chi2.sf(
    lr_stat,
    df_diff
)

# ------------------------------------------------------------
# 6. Extract exposure coefficients
# ------------------------------------------------------------

linear_coef = model_linear.params['log_exposure']
linear_p = model_linear.pvalues['log_exposure']

nonlinear_coef = model_nonlinear.params['log_exposure']
nonlinear_p = model_nonlinear.pvalues['log_exposure']

quadratic_coef = model_nonlinear.params['log_exposure_sq']
quadratic_p = model_nonlinear.pvalues['log_exposure_sq']

# ------------------------------------------------------------
# 7. Display results
# ------------------------------------------------------------

print("H2 — PRIOR EXPOSURES → RECALL")
print("=" * 75)

print(f"Observations used       : {len(h2_model_data):,}")
print(f"Total recall attempts   : {h2_model_data['session_seen'].sum():,}")

print("\nLINEAR EXPOSURE MODEL")
print("-" * 75)

print(f"Log exposure coefficient : {linear_coef:.6f}")
print(f"P-value                  : {linear_p:.10f}")

print("\nNON-LINEAR EXPOSURE MODEL")
print("-" * 75)

print(f"Log exposure coefficient      : {nonlinear_coef:.6f}")
print(f"P-value                       : {nonlinear_p:.10f}")
print(f"Log exposure² coefficient     : {quadratic_coef:.6f}")
print(f"Log exposure² P-value         : {quadratic_p:.10f}")

print("\nMODEL COMPARISON")
print("-" * 75)

print(f"Likelihood-ratio statistic : {lr_stat:.4f}")
print(f"Degrees of freedom         : {df_diff}")
print(f"P-value                    : {lr_p_value:.10f}")

print("\nINTERPRETATION")
print("-" * 75)

if lr_p_value < 0.05:
    print("The non-linear model significantly improves fit.")
    print("Evidence suggests the exposure → recall relationship is non-linear.")
else:
    print("The non-linear model does not significantly improve fit.")
    print("No strong evidence of non-linearity after controls.")

print("\nModel controls: lag_days + learning_language")
```

```python
# ============================================================
# H2 — MODEL-ADJUSTED PRIOR EXPOSURE → RECALL
# Final visualization before closing H2
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf

# ------------------------------------------------------------
# 1. Prepare modelling data
# ------------------------------------------------------------

h2_model = traces_cleaned[
    [
        'history_seen',
        'lag_days',
        'learning_language',
        'session_seen',
        'session_correct'
    ]
].copy()

# Remove invalid observations
h2_model = h2_model[
    (h2_model['history_seen'] >= 0) &
    (h2_model['lag_days'] >= 0) &
    (h2_model['session_seen'] > 0) &
    (h2_model['session_correct'] >= 0) &
    (h2_model['session_correct'] <= h2_model['session_seen']) &
    (h2_model['learning_language'].notna())
].copy()

# ------------------------------------------------------------
# 2. Create non-linear exposure variables
# ------------------------------------------------------------

h2_model['log_exposure'] = np.log1p(h2_model['history_seen'])
h2_model['log_exposure_sq'] = h2_model['log_exposure'] ** 2

h2_model['log_lag'] = np.log1p(h2_model['lag_days'])

# Recall proportion
h2_model['recall'] = (
    h2_model['session_correct'] /
    h2_model['session_seen']
)

# ------------------------------------------------------------
# 3. Fit non-linear logistic model
#    Controls:
#       - lag_days
#       - learning_language
# ------------------------------------------------------------

h2_model_fit = smf.glm(
    formula='''
        recall ~ log_exposure
        + log_exposure_sq
        + log_lag
        + C(learning_language)
    ''',
    data=h2_model,
    family=sm.families.Binomial(),
    freq_weights=h2_model['session_seen']
).fit()

# ------------------------------------------------------------
# 4. Extract coefficients
# ------------------------------------------------------------

b1 = h2_model_fit.params['log_exposure']
b2 = h2_model_fit.params['log_exposure_sq']

# Turning point in log(1 + exposure)
turning_log_exposure = -b1 / (2 * b2)

# Convert back to history_seen
turning_point = np.exp(turning_log_exposure) - 1

# ------------------------------------------------------------
# 5. Create exposure range for prediction
# ------------------------------------------------------------

# Focus on the meaningful exposure range while still
# allowing us to see the high-exposure decline.

exposure_values = np.unique(
    np.round(
        np.geomspace(1, 1000, 200)
    ).astype(int)
)

# Add the observed turning-point neighbourhood
exposure_values = np.unique(
    np.concatenate([
        exposure_values,
        np.arange(
            max(1, int(turning_point) - 5),
            int(turning_point) + 6
        )
    ])
)

# ------------------------------------------------------------
# 6. Hold other variables constant
# ------------------------------------------------------------

median_lag = h2_model['lag_days'].median()

# Most common learning language
reference_language = (
    h2_model['learning_language']
    .mode()
    .iloc[0]
)

prediction_data = pd.DataFrame({
    'history_seen': exposure_values,
    'log_exposure': np.log1p(exposure_values),
    'log_exposure_sq': np.log1p(exposure_values) ** 2,
    'log_lag': np.log1p(median_lag),
    'learning_language': reference_language
})

# ------------------------------------------------------------
# 7. Generate model-adjusted predictions
# ------------------------------------------------------------

prediction_data['predicted_recall'] = (
    h2_model_fit
    .predict(prediction_data)
    * 100
)

# ------------------------------------------------------------
# 8. Print final H2 results
# ------------------------------------------------------------

print("=" * 75)
print("H2 — MODEL-ADJUSTED PRIOR EXPOSURE → RECALL")
print("=" * 75)

print(f"Observations used          : {len(h2_model):,}")
print(f"Median lag held constant   : {median_lag:.2f} days")
print(f"Language held constant     : {reference_language}")

print("\nMODEL COEFFICIENTS")
print("-" * 75)
print(f"Log exposure coefficient   : {b1:.6f}")
print(f"Log exposure² coefficient  : {b2:.6f}")

print("\nESTIMATED TURNING POINT")
print("-" * 75)
print(f"Prior exposures            : {turning_point:.2f}")

# ------------------------------------------------------------
# 9. Compare predicted recall at key exposure levels
# ------------------------------------------------------------

key_exposures = [1, 5, 10, 20, 50, 100, 250, 500, 1000]

key_prediction = pd.DataFrame({
    'history_seen': key_exposures
})

key_prediction['log_exposure'] = np.log1p(
    key_prediction['history_seen']
)

key_prediction['log_exposure_sq'] = (
    key_prediction['log_exposure'] ** 2
)

key_prediction['log_lag'] = np.log1p(median_lag)
key_prediction['learning_language'] = reference_language

key_prediction['adjusted_recall_%'] = (
    h2_model_fit.predict(key_prediction) * 100
)

print("\nMODEL-ADJUSTED RECALL AT KEY EXPOSURE LEVELS")
print("-" * 75)

print(
    key_prediction[
        ['history_seen', 'adjusted_recall_%']
    ].to_string(
        index=False,
        formatters={
            'adjusted_recall_%': '{:.2f}'.format
        }
    )
)

# ------------------------------------------------------------
# 10. Plot adjusted exposure → recall curve
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    prediction_data['history_seen'],
    prediction_data['predicted_recall'],
    linewidth=2.5
)

plt.axvline(
    turning_point,
    linestyle='--',
    linewidth=1.5,
    label=f'Estimated turning point ≈ {turning_point:.1f} exposures'
)

plt.scatter(
    key_prediction['history_seen'],
    key_prediction['adjusted_recall_%'],
    s=35
)

plt.xscale('log')

plt.xlabel('Prior Exposures (history_seen)')
plt.ylabel('Model-Adjusted Recall Rate (%)')

plt.title(
    'Model-Adjusted Recall vs Prior Exposures\n'
    '(Controlling for Lag and Learning Language)'
)

plt.grid(alpha=0.25)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# H3 — WORD DIFFICULTY → RECALL
# PHASE 3: CONTROLLED ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import statsmodels.formula.api as smf

print("=" * 80)
print("H3 — WORD DIFFICULTY → RECALL")
print("PHASE 3: CONTROLLED ANALYSIS")
print("=" * 80)

# ------------------------------------------------------------
# 1. PREPARE DATA
# ------------------------------------------------------------

df_h3_model = word_difficulty_cleaned.copy()

required_cols = [
    'learning_language',
    'surface_form',
    'difficulty_rank_in_language',
    'item_recall_rate',
    'avg_prior_exposures',
    'n_traces'
]

missing_cols = [
    col for col in required_cols
    if col not in df_h3_model.columns
]

if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

df_h3_model = df_h3_model[required_cols].copy()

# Numeric conversion
numeric_cols = [
    'difficulty_rank_in_language',
    'item_recall_rate',
    'avg_prior_exposures',
    'n_traces'
]

for col in numeric_cols:
    df_h3_model[col] = pd.to_numeric(
        df_h3_model[col],
        errors='coerce'
    )

# Remove missing / invalid values
df_h3_model = df_h3_model.dropna(
    subset=required_cols
)

# Keep valid recall rates
df_h3_model = df_h3_model[
    (df_h3_model['item_recall_rate'] >= 0) &
    (df_h3_model['item_recall_rate'] <= 1) &
    (df_h3_model['avg_prior_exposures'] >= 0) &
    (df_h3_model['n_traces'] > 0)
].copy()

# ------------------------------------------------------------
# 2. TRANSFORM EXPOSURE
# ------------------------------------------------------------

df_h3_model['log_exposure'] = np.log1p(
    df_h3_model['avg_prior_exposures']
)

df_h3_model['log_exposure_sq'] = (
    df_h3_model['log_exposure'] ** 2
)

# Scale difficulty rank so coefficient is easier to interpret
# One unit = 100 rank positions
df_h3_model['difficulty_rank_100'] = (
    df_h3_model['difficulty_rank_in_language'] / 100
)

print("\nMODEL DATA")
print("-" * 80)
print("Word-language combinations :", f"{len(df_h3_model):,}")
print(
    "Learning languages         :",
    df_h3_model['learning_language'].nunique()
)

print(
    "Difficulty rank range      :",
    f"{df_h3_model['difficulty_rank_in_language'].min():.0f}"
    f" → "
    f"{df_h3_model['difficulty_rank_in_language'].max():.0f}"
)

# ------------------------------------------------------------
# 3. CONTROLLED MODEL
# ------------------------------------------------------------
#
# Outcome:
#   item_recall_rate
#
# Main variable:
#   difficulty_rank_in_language
#
# Controls:
#   log(avg_prior_exposures)
#   log(avg_prior_exposures)^2
#   learning_language
#
# Weights:
#   n_traces
#
# Higher difficulty rank = easier word.
# Therefore a POSITIVE difficulty coefficient means:
# easier words → higher recall.
# ------------------------------------------------------------

model = smf.wls(
    formula="""
        item_recall_rate
        ~ difficulty_rank_100
        + log_exposure
        + log_exposure_sq
        + C(learning_language)
    """,
    data=df_h3_model,
    weights=df_h3_model['n_traces']
).fit()

# ------------------------------------------------------------
# 4. EXTRACT DIFFICULTY RESULT
# ------------------------------------------------------------

difficulty_coef = model.params['difficulty_rank_100']
difficulty_p = model.pvalues['difficulty_rank_100']

difficulty_ci = model.conf_int().loc[
    'difficulty_rank_100'
]

print("\n" + "=" * 80)
print("H3 — CONTROLLED DIFFICULTY EFFECT")
print("=" * 80)

print(
    f"\nDifficulty coefficient : {difficulty_coef:.6f}"
)

print(
    f"P-value                : {difficulty_p:.10f}"
)

print(
    f"95% CI                 : "
    f"{difficulty_ci.iloc[0]:.6f}"
    f" to "
    f"{difficulty_ci.iloc[1]:.6f}"
)

print(
    f"\nModel R²               : {model.rsquared:.4f}"
)

# Convert coefficient into percentage points
difficulty_effect_pp = difficulty_coef * 100

print(
    "\nRecall effect per +100 difficulty-rank positions:"
)

print(
    f"{difficulty_effect_pp:+.3f} percentage points"
)

# ------------------------------------------------------------
# 5. EXPOSURE COEFFICIENTS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CONTROL VARIABLES")
print("=" * 80)

print("\nExposure effects:")

print(
    f"Log exposure coefficient   : "
    f"{model.params['log_exposure']:.6f}"
)

print(
    f"Log exposure² coefficient  : "
    f"{model.params['log_exposure_sq']:.6f}"
)

print(
    f"Log exposure p-value       : "
    f"{model.pvalues['log_exposure']:.10f}"
)

print(
    f"Log exposure² p-value      : "
    f"{model.pvalues['log_exposure_sq']:.10f}"
)

# ------------------------------------------------------------
# 6. LANGUAGE EFFECTS
# ------------------------------------------------------------

print("\nLanguage controls:")

language_rows = []

for term in model.params.index:

    if term.startswith("C(learning_language)"):

        language_rows.append({
            'term': term,
            'coefficient': model.params[term],
            'p_value': model.pvalues[term]
        })

language_effects = pd.DataFrame(language_rows)

if len(language_effects) > 0:
    print(
        language_effects.to_string(index=False)
    )

# ------------------------------------------------------------
# 7. HYPOTHESIS DECISION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("H3 HYPOTHESIS DECISION")
print("=" * 80)

print("""
H0: Word difficulty has no effect on recall
H1: More difficult words have lower recall
""")

if difficulty_p < 0.05:

    print("Decision: REJECT H0")

    if difficulty_coef > 0:
        print(
            "Evidence supports H1: "
            "easier words have higher recall, "
            "even after controlling for prior exposure "
            "and learning language."
        )
    else:
        print(
            "Difficulty remains statistically significant, "
            "but the direction is opposite to the hypothesis."
        )

else:

    print("Decision: FAIL TO REJECT H0")

    print(
        "After controlling for prior exposure and language, "
        "the evidence is insufficient to establish an "
        "independent difficulty effect."
    )

# ------------------------------------------------------------
# 8. FINAL MODEL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("H3 PHASE 3 SUMMARY")
print("=" * 80)

print(
    f"Words analysed              : {len(df_h3_model):,}"
)

print(
    f"Difficulty effect / 100 ranks: "
    f"{difficulty_effect_pp:+.3f} pp"
)

print(
    f"Difficulty p-value           : "
    f"{difficulty_p:.10f}"
)

print(
    f"Model R²                     : "
    f"{model.rsquared:.4f}"
)
```

```python
# ================================================================
# H4 — ADDITIONAL EXPOSURES HAVE DIMINISHING RETURNS
# PHASE 1: MARGINAL RECALL GAINS
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 80)
print("H4 — ADDITIONAL EXPOSURES → DIMINISHING RETURNS")
print("PHASE 1: MARGINAL RECALL GAINS")
print("=" * 80)

# ------------------------------------------------
# 1. Create exposure bands
# ------------------------------------------------

exposure_bins = [0, 1, 2, 4, 9, 19, 49, 99, 249, np.inf]

exposure_labels = [
    "1",
    "2",
    "3–4",
    "5–9",
    "10–19",
    "20–49",
    "50–99",
    "100–249",
    "250+"
]

h4_data = traces_cleaned.copy()

h4_data = h4_data[
    h4_data["history_seen"].notna() &
    h4_data["session_seen"].notna() &
    h4_data["session_correct"].notna()
].copy()

# history_seen = number of prior exposures
h4_data["exposure_group"] = pd.cut(
    h4_data["history_seen"],
    bins=exposure_bins,
    labels=exposure_labels,
    include_lowest=True
)

# ------------------------------------------------
# 2. Calculate recall by exposure group
# ------------------------------------------------

exposure_summary = (
    h4_data
    .groupby("exposure_group", observed=False)
    .agg(
        total_attempts=("session_seen", "sum"),
        total_correct=("session_correct", "sum"),
        n_traces=("history_seen", "size")
    )
    .reset_index()
)

exposure_summary["recall_%"] = (
    exposure_summary["total_correct"] /
    exposure_summary["total_attempts"] * 100
)

# ------------------------------------------------
# 3. Calculate change versus previous exposure group
# ------------------------------------------------

exposure_summary["recall_change_pp"] = (
    exposure_summary["recall_%"].diff()
)

# ------------------------------------------------
# 4. Display results
# ------------------------------------------------

print("\nEXPOSURE → RECALL SUMMARY")
print("-" * 80)

display(
    exposure_summary[
        [
            "exposure_group",
            "n_traces",
            "total_attempts",
            "recall_%",
            "recall_change_pp"
        ]
    ].round(2)
)

# ------------------------------------------------
# 5. Plot recall and marginal change
# ------------------------------------------------

fig, ax = plt.subplots(figsize=(11, 6))

x = np.arange(len(exposure_summary))

ax.plot(
    x,
    exposure_summary["recall_%"],
    marker="o",
    linewidth=2
)

ax.set_xticks(x)
ax.set_xticklabels(
    exposure_summary["exposure_group"],
    rotation=45,
    ha="right"
)

ax.set_xlabel("Prior Exposures")
ax.set_ylabel("Recall Rate (%)")
ax.set_title(
    "Recall Rate by Prior Exposure Group\n"
    "Testing for Diminishing Returns"
)

ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

# ------------------------------------------------
# 6. Identify maximum recall
# ------------------------------------------------

peak_idx = exposure_summary["recall_%"].idxmax()

peak_group = exposure_summary.loc[peak_idx, "exposure_group"]
peak_recall = exposure_summary.loc[peak_idx, "recall_%"]

print("\nKEY FINDING")
print("=" * 80)
print(f"Highest observed recall group : {peak_group}")
print(f"Recall at peak               : {peak_recall:.2f}%")
```

```python
# ================================================================
# H4 — ADDITIONAL EXPOSURES → DIMINISHING RETURNS
# PHASE 2: CONTROLLED NON-LINEAR TEST
# ================================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf

print("=" * 80)
print("H4 — ADDITIONAL EXPOSURES → DIMINISHING RETURNS")
print("PHASE 2: CONTROLLED NON-LINEAR TEST")
print("=" * 80)

# ------------------------------------------------
# 1. Prepare data
# ------------------------------------------------

h4_model = traces_cleaned.copy()

h4_model = h4_model[
    h4_model["history_seen"].notna() &
    h4_model["lag_days"].notna() &
    h4_model["learning_language"].notna() &
    h4_model["session_seen"].notna() &
    h4_model["session_correct"].notna()
].copy()

# Remove impossible values
h4_model = h4_model[
    (h4_model["history_seen"] >= 1) &
    (h4_model["session_seen"] > 0) &
    (h4_model["session_correct"] >= 0) &
    (h4_model["session_correct"] <= h4_model["session_seen"])
].copy()

# Log exposure
h4_model["log_exposure"] = np.log1p(h4_model["history_seen"])

# Log lag
h4_model["log_lag"] = np.log1p(h4_model["lag_days"])

# ------------------------------------------------
# 2. Fit linear exposure model
# ------------------------------------------------

linear_model = smf.glm(
    formula="""
        calculated_recall
        ~ log_exposure
        + log_lag
        + C(learning_language)
    """,
    data=h4_model,
    family=sm.families.Binomial(),
    freq_weights=h4_model["session_seen"]
).fit()

# ------------------------------------------------
# 3. Fit non-linear exposure model
# ------------------------------------------------

h4_model["log_exposure_sq"] = (
    h4_model["log_exposure"] ** 2
)

nonlinear_model = smf.glm(
    formula="""
        calculated_recall
        ~ log_exposure
        + log_exposure_sq
        + log_lag
        + C(learning_language)
    """,
    data=h4_model,
    family=sm.families.Binomial(),
    freq_weights=h4_model["session_seen"]
).fit()

# ------------------------------------------------
# 4. Compare models
# ------------------------------------------------

lr_stat = (
    linear_model.deviance -
    nonlinear_model.deviance
)

df_diff = (
    linear_model.df_resid -
    nonlinear_model.df_resid
)

from scipy.stats import chi2

lr_pvalue = chi2.sf(lr_stat, df_diff)

# ------------------------------------------------
# 5. Extract coefficients
# ------------------------------------------------

linear_coef = linear_model.params["log_exposure"]
linear_p = linear_model.pvalues["log_exposure"]

nonlinear_coef = nonlinear_model.params["log_exposure"]
nonlinear_sq_coef = nonlinear_model.params["log_exposure_sq"]

nonlinear_p = nonlinear_model.pvalues["log_exposure"]
nonlinear_sq_p = nonlinear_model.pvalues["log_exposure_sq"]

# ------------------------------------------------
# 6. Calculate turning point
# ------------------------------------------------

turning_log_exposure = (
    -nonlinear_coef /
    (2 * nonlinear_sq_coef)
)

turning_exposure = (
    np.exp(turning_log_exposure) - 1
)

# ------------------------------------------------
# 7. Print results
# ------------------------------------------------

print("\nMODEL DATA")
print("-" * 80)
print(f"Observations used        : {len(h4_model):,}")
print(f"Total recall attempts    : {h4_model['session_seen'].sum():,}")
print(
    f"Exposure range           : "
    f"{h4_model['history_seen'].min():.0f} → "
    f"{h4_model['history_seen'].max():.0f}"
)

print("\nLINEAR EXPOSURE MODEL")
print("-" * 80)
print(f"Log exposure coefficient : {linear_coef:.6f}")
print(f"P-value                  : {linear_p:.10f}")

print("\nNON-LINEAR EXPOSURE MODEL")
print("-" * 80)
print(f"Log exposure coefficient     : {nonlinear_coef:.6f}")
print(f"P-value                      : {nonlinear_p:.10f}")
print(
    f"Log exposure² coefficient    : "
    f"{nonlinear_sq_coef:.6f}"
)
print(
    f"Log exposure² P-value        : "
    f"{nonlinear_sq_p:.10f}"
)

print("\nMODEL COMPARISON")
print("-" * 80)
print(f"Likelihood-ratio statistic : {lr_stat:.4f}")
print(f"Degrees of freedom         : {df_diff}")
print(f"P-value                    : {lr_pvalue:.10f}")

print("\nESTIMATED TURNING POINT")
print("-" * 80)
print(
    f"Prior exposures            : "
    f"{turning_exposure:.2f}"
)

print("\nH4 HYPOTHESIS")
print("-" * 80)
print("H0: Additional exposures provide constant marginal benefit")
print("H1: Marginal recall benefit decreases as exposures increase")

print("\nDECISION")
print("-" * 80)

if lr_pvalue < 0.05 and nonlinear_sq_coef < 0:
    print("Reject H0")
    print(
        "Evidence supports diminishing returns from "
        "additional exposures."
    )
else:
    print("Do not reject H0")
    print(
        "Evidence is insufficient to establish "
        "diminishing returns."
    )
```

```python
# ================================================================
# H4 — ADDITIONAL EXPOSURES → DIMINISHING RETURNS
# PHASE 3: PRACTICAL EXPOSURE THRESHOLD
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf

print("=" * 80)
print("H4 — ADDITIONAL EXPOSURES → DIMINISHING RETURNS")
print("PHASE 3: PRACTICAL EXPOSURE THRESHOLD")
print("=" * 80)

# ------------------------------------------------
# 1. Prepare model data
# ------------------------------------------------

h4_final = traces_cleaned.copy()

h4_final = h4_final[
    h4_final["history_seen"].notna() &
    h4_final["lag_days"].notna() &
    h4_final["learning_language"].notna() &
    h4_final["session_seen"].notna() &
    h4_final["session_correct"].notna()
].copy()

h4_final = h4_final[
    (h4_final["history_seen"] >= 1) &
    (h4_final["session_seen"] > 0) &
    (h4_final["session_correct"] >= 0) &
    (h4_final["session_correct"] <= h4_final["session_seen"])
].copy()

# Log transformations
h4_final["log_exposure"] = np.log1p(
    h4_final["history_seen"]
)

h4_final["log_exposure_sq"] = (
    h4_final["log_exposure"] ** 2
)

h4_final["log_lag"] = np.log1p(
    h4_final["lag_days"]
)

# ------------------------------------------------
# 2. Fit final controlled model
# ------------------------------------------------

h4_model = smf.glm(
    formula="""
        calculated_recall
        ~ log_exposure
        + log_exposure_sq
        + log_lag
        + C(learning_language)
    """,
    data=h4_final,
    family=sm.families.Binomial(),
    freq_weights=h4_final["session_seen"]
).fit()

# ------------------------------------------------
# 3. Create exposure levels to evaluate
# ------------------------------------------------

exposure_levels = [
    1, 2, 5, 10, 20, 50, 100, 250, 500, 1000
]

# Hold lag and language constant
median_lag = h4_final["lag_days"].median()
reference_language = "en"

prediction_data = pd.DataFrame({
    "history_seen": exposure_levels
})

prediction_data["log_exposure"] = np.log1p(
    prediction_data["history_seen"]
)

prediction_data["log_exposure_sq"] = (
    prediction_data["log_exposure"] ** 2
)

prediction_data["log_lag"] = np.log1p(median_lag)

prediction_data["learning_language"] = reference_language

# ------------------------------------------------
# 4. Generate adjusted recall predictions
# ------------------------------------------------

prediction_data["adjusted_recall_%"] = (
    h4_model.predict(prediction_data) * 100
)

# ------------------------------------------------
# 5. Calculate incremental change
# ------------------------------------------------

prediction_data["change_vs_previous_pp"] = (
    prediction_data["adjusted_recall_%"].diff()
)

# ------------------------------------------------
# 6. Calculate gain relative to 1 exposure
# ------------------------------------------------

baseline = prediction_data.loc[
    prediction_data["history_seen"] == 1,
    "adjusted_recall_%"
].iloc[0]

prediction_data["gain_vs_1_exposure_pp"] = (
    prediction_data["adjusted_recall_%"] - baseline
)

# ------------------------------------------------
# 7. Display practical table
# ------------------------------------------------

print("\nMODEL-ADJUSTED RECALL BY EXPOSURE LEVEL")
print("-" * 80)

display(
    prediction_data[
        [
            "history_seen",
            "adjusted_recall_%",
            "change_vs_previous_pp",
            "gain_vs_1_exposure_pp"
        ]
    ].round(3)
)

# ------------------------------------------------
# 8. Identify maximum predicted recall
# ------------------------------------------------

peak_idx = prediction_data[
    "adjusted_recall_%"
].idxmax()

peak_exposure = prediction_data.loc[
    peak_idx, "history_seen"
]

peak_recall = prediction_data.loc[
    peak_idx, "adjusted_recall_%"
]

print("\nPEAK MODEL-ADJUSTED RECALL")
print("-" * 80)
print(f"Exposure level : {peak_exposure}")
print(f"Recall         : {peak_recall:.3f}%")

# ------------------------------------------------
# 9. Plot adjusted recall curve
# ------------------------------------------------

plt.figure(figsize=(11, 6))

plt.plot(
    prediction_data["history_seen"],
    prediction_data["adjusted_recall_%"],
    marker="o"
)

plt.xscale("log")

plt.xlabel("Prior Exposures (log scale)")
plt.ylabel("Model-Adjusted Recall (%)")

plt.title(
    "Model-Adjusted Recall vs Prior Exposures\n"
    "Identifying the Practical Exposure Threshold"
)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ------------------------------------------------
# 10. Business interpretation
# ------------------------------------------------

print("\nH4 BUSINESS INTERPRETATION")
print("=" * 80)

print(
    f"Estimated statistical turning point : "
    f"approximately 10 exposures"
)

print(
    "Recall improves during the early exposure stage, "
    "then the marginal benefit becomes progressively smaller."
)

print(
    "After the turning point, additional exposure does not "
    "produce sustained recall improvement."
)

print(
    "\nImportant: this analysis identifies an exposure threshold "
    "associated with recall, not a causal prescription for the "
    "number of reviews Duolingo should schedule."
)
```

```python
# ================================================================
# H6 — LEARNING LANGUAGE → RECALL
# PHASE 2: STATISTICAL SIGNIFICANCE TEST
# ================================================================

import numpy as np
from scipy.stats import chi2_contingency

print("=" * 80)
print("H6 — LEARNING LANGUAGE → RECALL")
print("PHASE 2: STATISTICAL SIGNIFICANCE TEST")
print("=" * 80)

# ------------------------------------------------
# 1. Build language × recall outcome table
# ------------------------------------------------

h6_test = (
    traces_cleaned
    .groupby("learning_language")
    .agg(
        total_attempts=("session_seen", "sum"),
        total_correct=("session_correct", "sum")
    )
    .reset_index()
)

h6_test["total_incorrect"] = (
    h6_test["total_attempts"] -
    h6_test["total_correct"]
)

# Contingency table:
# rows    = learning languages
# columns = correct / incorrect

contingency_table = h6_test[
    ["total_correct", "total_incorrect"]
].values

# ------------------------------------------------
# 2. Chi-square test of independence
# ------------------------------------------------

chi2_stat, p_value, degrees_freedom, expected = chi2_contingency(
    contingency_table
)

# ------------------------------------------------
# 3. Effect size — Cramer's V
# ------------------------------------------------

n_total = contingency_table.sum()

n_rows, n_cols = contingency_table.shape

cramers_v = np.sqrt(
    chi2_stat /
    (n_total * min(n_rows - 1, n_cols - 1))
)

# ------------------------------------------------
# 4. Display test table
# ------------------------------------------------

print("\nOBSERVED RECALL OUTCOMES BY LANGUAGE")
print("-" * 80)

display(
    h6_test[
        [
            "learning_language",
            "total_attempts",
            "total_correct",
            "total_incorrect"
        ]
    ].sort_values(
        "total_attempts",
        ascending=False
    )
)

print("\n" + "=" * 80)
print("CHI-SQUARE TEST")
print("=" * 80)

print(f"Chi-square statistic : {chi2_stat:,.2f}")
print(f"Degrees of freedom  : {degrees_freedom}")
print(f"P-value             : {p_value:.10f}")

print("\n" + "=" * 80)
print("EFFECT SIZE")
print("=" * 80)

print(f"Cramer's V           : {cramers_v:.6f}")

# ------------------------------------------------
# 5. Hypothesis decision
# ------------------------------------------------

alpha = 0.05

print("\n" + "=" * 80)
print("HYPOTHESES")
print("=" * 80)

print("H0: Word Recall Rate is the same across learning languages")
print("H1: Word Recall Rate differs across learning languages")

print("\nDecision rule")
print("-" * 80)
print(f"Significance level (α): {alpha}")

if p_value < alpha:
    print("Result: Reject H0")
    print("Evidence indicates that recall differs significantly across languages.")
else:
    print("Result: Fail to reject H0")
    print("Insufficient evidence that recall differs across languages.")

# ------------------------------------------------
# 6. Interpretation guide
# ------------------------------------------------

print("\n" + "=" * 80)
print("INTERPRETATION")
print("=" * 80)

print(
    "Cramer's V measures the magnitude of the association between "
    "learning language and recall outcome."
)

print(
    "The statistical test establishes whether language-level recall "
    "differences are unlikely to be due to random variation."
)

print("\n" + "=" * 80)
print("H6 PHASE 2 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H6 — LEARNING LANGUAGE → RECALL
# PHASE 3: CONTROLLED LANGUAGE EFFECT
# ================================================================

import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

print("=" * 80)
print("H6 — LEARNING LANGUAGE → RECALL")
print("PHASE 3: CONTROLLED LANGUAGE EFFECT")
print("=" * 80)

# ------------------------------------------------
# 1. Prepare model data
# ------------------------------------------------

h6_model = traces_cleaned[
    [
        "session_seen",
        "session_correct",
        "learning_language",
        "lag_days",
        "history_seen"
    ]
].copy()

# Keep valid observations only
h6_model = h6_model.dropna(
    subset=[
        "session_seen",
        "session_correct",
        "learning_language",
        "lag_days",
        "history_seen"
    ]
)

# Remove rows with zero attempts
h6_model = h6_model[
    h6_model["session_seen"] > 0
].copy()

# ------------------------------------------------
# 2. Create model variables
# ------------------------------------------------

h6_model["log_lag"] = np.log1p(
    h6_model["lag_days"].clip(lower=0)
)

h6_model["log_exposure"] = np.log1p(
    h6_model["history_seen"].clip(lower=0)
)

h6_model["incorrect"] = (
    h6_model["session_seen"] -
    h6_model["session_correct"]
)

print("\nMODEL DATA")
print("-" * 80)

print(f"Observations used : {len(h6_model):,}")
print(
    f"Total recall attempts : "
    f"{h6_model['session_seen'].sum():,}"
)

print(
    f"Languages : "
    f"{h6_model['learning_language'].nunique()}"
)

# ------------------------------------------------
# 3. Controlled logistic regression
# ------------------------------------------------
#
# English is the reference language.
#
# The model estimates:
#
# Recall ~ language + lag + exposure
#
# ------------------------------------------------

formula = """
session_correct + incorrect
~ C(learning_language, Treatment(reference='en'))
+ log_lag
+ log_exposure
"""

h6_model_fit = smf.glm(
    formula=formula,
    data=h6_model,
    family=__import__("statsmodels").api.families.Binomial(),
    freq_weights=h6_model["session_seen"]
).fit()

# ------------------------------------------------
# 4. Extract language coefficients
# ------------------------------------------------

language_terms = [
    term for term in h6_model_fit.params.index
    if "learning_language" in term
]

language_results = pd.DataFrame({
    "term": language_terms,
    "coefficient": [
        h6_model_fit.params[t]
        for t in language_terms
    ],
    "odds_ratio": [
        np.exp(h6_model_fit.params[t])
        for t in language_terms
    ],
    "p_value": [
        h6_model_fit.pvalues[t]
        for t in language_terms
    ]
})

# Confidence intervals
language_results["ci_lower"] = [
    np.exp(h6_model_fit.conf_int().loc[t, 0])
    for t in language_terms
]

language_results["ci_upper"] = [
    np.exp(h6_model_fit.conf_int().loc[t, 1])
    for t in language_terms
]

print("\n" + "=" * 80)
print("CONTROLLED LANGUAGE EFFECTS")
print("=" * 80)

print("\nReference language: English")

display(
    language_results.sort_values("p_value")
)

# ------------------------------------------------
# 5. Model controls
# ------------------------------------------------

print("\n" + "=" * 80)
print("CONTROL VARIABLES")
print("=" * 80)

for term in ["log_lag", "log_exposure"]:

    print(
        f"{term:<20} "
        f"coefficient = {h6_model_fit.params[term]:.6f}   "
        f"p-value = {h6_model_fit.pvalues[term]:.10f}"
    )

# ------------------------------------------------
# 6. Hypothesis decision
# ------------------------------------------------

alpha = 0.05

significant_languages = language_results[
    language_results["p_value"] < alpha
]

print("\n" + "=" * 80)
print("H6 HYPOTHESIS DECISION")
print("=" * 80)

print(
    "H0: Learning language has no independent effect on recall "
    "after controlling for lag and prior exposure"
)

print(
    "H1: Learning language has an independent effect on recall "
    "after controlling for lag and prior exposure"
)

print("\nDecision")

if len(significant_languages) > 0:
    print("Reject H0")
    print(
        "Evidence suggests that at least one learning language "
        "has a statistically significant recall difference "
        "relative to English after controlling for lag and exposure."
    )
else:
    print("Fail to reject H0")
    print(
        "No learning language shows a statistically significant "
        "difference from English after controlling for lag and exposure."
    )

# ------------------------------------------------
# 7. Interpretation guide
# ------------------------------------------------

print("\n" + "=" * 80)
print("INTERPRETATION GUIDE")
print("=" * 80)

print(
    "Odds Ratio > 1  → higher odds of correct recall than English"
)

print(
    "Odds Ratio < 1  → lower odds of correct recall than English"
)

print(
    "The controlled language effects should be interpreted "
    "together with their p-values and confidence intervals."
)

print("\n" + "=" * 80)
print("H6 PHASE 3 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H8 — LANGUAGE PAIR → RECALL
# PHASE 3: CONTROLLED LANGUAGE-PAIR EFFECT
# ================================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf

print("=" * 80)
print("H8 — LANGUAGE PAIR → RECALL")
print("PHASE 3: CONTROLLED LANGUAGE-PAIR EFFECT")
print("=" * 80)

# ------------------------------------------------
# 1. Prepare model data
# ------------------------------------------------

h8_model = traces_cleaned[
    [
        "session_seen",
        "session_correct",
        "ui_language",
        "learning_language",
        "lag_days",
        "history_seen"
    ]
].copy()

h8_model = h8_model.dropna(
    subset=[
        "session_seen",
        "session_correct",
        "ui_language",
        "learning_language",
        "lag_days",
        "history_seen"
    ]
)

# Keep valid recall attempts
h8_model = h8_model[
    h8_model["session_seen"] > 0
].copy()

# ------------------------------------------------
# 2. Create language pair
# ------------------------------------------------

h8_model["language_pair"] = (
    h8_model["ui_language"]
    + " → "
    + h8_model["learning_language"]
)

# ------------------------------------------------
# 3. Create model variables
# ------------------------------------------------

h8_model["log_lag"] = np.log1p(
    h8_model["lag_days"].clip(lower=0)
)

h8_model["log_exposure"] = np.log1p(
    h8_model["history_seen"].clip(lower=0)
)

h8_model["incorrect"] = (
    h8_model["session_seen"]
    - h8_model["session_correct"]
)

# ------------------------------------------------
# 4. Check available language pairs
# ------------------------------------------------

print("\nMODEL DATA")
print("-" * 80)

print(
    f"Observations used       : {len(h8_model):,}"
)

print(
    f"Total recall attempts   : "
    f"{h8_model['session_seen'].sum():,}"
)

print(
    f"Language pairs          : "
    f"{h8_model['language_pair'].nunique()}"
)

print("\nLanguage pairs:")
print(
    sorted(
        h8_model["language_pair"].unique()
    )
)

# ------------------------------------------------
# 5. Controlled logistic regression
# ------------------------------------------------
#
# en → fr is used as the reference pair.
#
# Model:
#
# Recall ~ Language Pair + Lag + Prior Exposure
#
# ------------------------------------------------

formula = """
session_correct + incorrect
~ C(
    language_pair,
    Treatment(reference='en → fr')
)
+ log_lag
+ log_exposure
"""

h8_fit = smf.glm(
    formula=formula,
    data=h8_model,
    family=sm.families.Binomial(),
    freq_weights=h8_model["session_seen"]
).fit()

# ------------------------------------------------
# 6. Extract language-pair effects
# ------------------------------------------------

pair_terms = [
    term
    for term in h8_fit.params.index
    if "language_pair" in term
]

pair_results = pd.DataFrame({
    "term": pair_terms,
    "coefficient": [
        h8_fit.params[t]
        for t in pair_terms
    ],
    "odds_ratio": [
        np.exp(h8_fit.params[t])
        for t in pair_terms
    ],
    "p_value": [
        h8_fit.pvalues[t]
        for t in pair_terms
    ]
})

# Confidence intervals
conf = h8_fit.conf_int()

pair_results["ci_lower"] = [
    np.exp(conf.loc[t, 0])
    for t in pair_terms
]

pair_results["ci_upper"] = [
    np.exp(conf.loc[t, 1])
    for t in pair_terms
]

# ------------------------------------------------
# 7. Display results
# ------------------------------------------------

print("\n" + "=" * 80)
print("CONTROLLED LANGUAGE-PAIR EFFECTS")
print("=" * 80)

print("\nReference language pair: en → fr")

display(
    pair_results.sort_values(
        "odds_ratio",
        ascending=False
    )
)

# ------------------------------------------------
# 8. Control variables
# ------------------------------------------------

print("\n" + "=" * 80)
print("CONTROL VARIABLES")
print("=" * 80)

print(
    f"log_lag              coefficient = "
    f"{h8_fit.params['log_lag']:.6f}   "
    f"p-value = "
    f"{h8_fit.pvalues['log_lag']:.10f}"
)

print(
    f"log_exposure         coefficient = "
    f"{h8_fit.params['log_exposure']:.6f}   "
    f"p-value = "
    f"{h8_fit.pvalues['log_exposure']:.10f}"
)

# ------------------------------------------------
# 9. Hypothesis decision
# ------------------------------------------------

alpha = 0.05

significant_pairs = pair_results[
    pair_results["p_value"] < alpha
]

print("\n" + "=" * 80)
print("H8 HYPOTHESIS DECISION")
print("=" * 80)

print(
    "H0: Language pair has no independent effect "
    "on recall after controlling for lag and prior exposure"
)

print(
    "H1: Language pair has an independent effect "
    "on recall after controlling for lag and prior exposure"
)

print("\nDecision")

if len(significant_pairs) > 0:
    print("Reject H0")
    print(
        "Evidence suggests that at least one language pair "
        "differs significantly from en → fr after controlling "
        "for lag and prior exposure."
    )
else:
    print("Fail to reject H0")
    print(
        "No language pair shows a statistically significant "
        "difference from en → fr after controlling for lag "
        "and prior exposure."
    )

# ------------------------------------------------
# 10. Interpretation guide
# ------------------------------------------------

print("\n" + "=" * 80)
print("INTERPRETATION GUIDE")
print("=" * 80)

print(
    "Odds Ratio > 1 → higher odds of correct recall "
    "than en → fr"
)

print(
    "Odds Ratio < 1 → lower odds of correct recall "
    "than en → fr"
)

print(
    "95% CI excluding 1 indicates a statistically "
    "significant pair difference at α = 0.05."
)

print(
    "These are associations, not causal effects."
)

print("\n" + "=" * 80)
print("H8 PHASE 3 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H9 — PART OF SPEECH → RECALL
# PHASE 1: POS-LEVEL DIFFERENCES
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 80)
print("H9 — PART OF SPEECH → RECALL")
print("PHASE 1: POS-LEVEL DIFFERENCES")
print("=" * 80)

# ------------------------------------------------
# 1. Prepare data
# ------------------------------------------------

h9_data = traces_cleaned[
    [
        "pos",
        "surface_form",
        "session_seen",
        "session_correct"
    ]
].copy()

h9_data = h9_data.dropna(
    subset=[
        "pos",
        "surface_form",
        "session_seen",
        "session_correct"
    ]
)

# Keep valid recall attempts
h9_data = h9_data[
    h9_data["session_seen"] > 0
].copy()

# Incorrect recalls
h9_data["incorrect"] = (
    h9_data["session_seen"]
    - h9_data["session_correct"]
)

# ------------------------------------------------
# 2. Count unique words by POS
# ------------------------------------------------

pos_word_counts = (
    h9_data
    .groupby("pos")["surface_form"]
    .nunique()
    .reset_index(name="n_words")
)

# ------------------------------------------------
# 3. Aggregate recall by POS
# ------------------------------------------------

pos_summary = (
    h9_data
    .groupby("pos")
    .agg(
        n_traces=("pos", "size"),
        total_attempts=("session_seen", "sum"),
        total_correct=("session_correct", "sum"),
        total_incorrect=("incorrect", "sum")
    )
    .reset_index()
)

# Add unique word count
pos_summary = pos_summary.merge(
    pos_word_counts,
    on="pos",
    how="left"
)

# ------------------------------------------------
# 4. Calculate recall metrics
# ------------------------------------------------

pos_summary["recall_%"] = (
    pos_summary["total_correct"]
    / pos_summary["total_attempts"]
    * 100
)

pos_summary["incorrect_%"] = (
    pos_summary["total_incorrect"]
    / pos_summary["total_attempts"]
    * 100
)

# ------------------------------------------------
# 5. Reliability filter
# ------------------------------------------------
#
# Keep POS categories with at least 20 unique words.
# This removes extremely small / special POS categories.
# ------------------------------------------------

MIN_WORDS = 20

pos_summary_filtered = pos_summary[
    pos_summary["n_words"] >= MIN_WORDS
].copy()

pos_summary_filtered = (
    pos_summary_filtered
    .sort_values("recall_%")
    .reset_index(drop=True)
)

# ------------------------------------------------
# 6. Display full POS summary
# ------------------------------------------------

print("\nPOS RECALL SUMMARY")
print("-" * 80)

display(
    pos_summary_filtered[
        [
            "pos",
            "n_words",
            "n_traces",
            "total_attempts",
            "total_correct",
            "total_incorrect",
            "recall_%",
            "incorrect_%"
        ]
    ]
)

# ------------------------------------------------
# 7. Highest and lowest recall POS
# ------------------------------------------------

lowest = pos_summary_filtered.loc[
    pos_summary_filtered["recall_%"].idxmin()
]

highest = pos_summary_filtered.loc[
    pos_summary_filtered["recall_%"].idxmax()
]

recall_gap = (
    highest["recall_%"]
    - lowest["recall_%"]
)

print("\n" + "=" * 80)
print("KEY FINDINGS")
print("=" * 80)

print(
    f"Lowest recall POS  : "
    f"{lowest['pos']} → {lowest['recall_%']:.2f}%"
)

print(
    f"Highest recall POS : "
    f"{highest['pos']} → {highest['recall_%']:.2f}%"
)

print(
    f"Recall gap         : "
    f"{recall_gap:.2f} percentage points"
)

print(
    f"POS categories tested : "
    f"{len(pos_summary_filtered)}"
)

print(
    f"Minimum words per POS : "
    f"{MIN_WORDS}"
)

# ------------------------------------------------
# 8. Recall change relative to lowest POS
# ------------------------------------------------

pos_summary_filtered["recall_change_vs_lowest_pp"] = (
    pos_summary_filtered["recall_%"]
    - lowest["recall_%"]
)

print("\nRECALL CHANGE VS LOWEST-RECALL POS")
print("-" * 80)

display(
    pos_summary_filtered[
        [
            "pos",
            "n_words",
            "recall_%",
            "recall_change_vs_lowest_pp"
        ]
    ]
)

# ------------------------------------------------
# 9. Visualization
# ------------------------------------------------

plot_data = pos_summary_filtered.sort_values(
    "recall_%"
)

plt.figure(figsize=(10, 6))

bars = plt.barh(
    plot_data["pos"],
    plot_data["recall_%"]
)

plt.xlabel("Word Recall Rate (%)")
plt.ylabel("Part of Speech")
plt.title(
    "Word Recall Rate by Part of Speech\n"
    f"(POS categories with ≥{MIN_WORDS} words)"
)

plt.xlim(
    plot_data["recall_%"].min() - 0.5,
    plot_data["recall_%"].max() + 0.5
)

for bar, value in zip(
    bars,
    plot_data["recall_%"]
):
    plt.text(
        bar.get_width() + 0.03,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.2f}%",
        va="center"
    )

plt.tight_layout()
plt.show()

print("\n" + "=" * 80)
print("H9 PHASE 1 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H9 — PART OF SPEECH → RECALL
# PHASE 2: STATISTICAL SIGNIFICANCE TEST
# ================================================================

import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

print("=" * 80)
print("H9 — PART OF SPEECH → RECALL")
print("PHASE 2: STATISTICAL SIGNIFICANCE TEST")
print("=" * 80)

# ------------------------------------------------
# 1. Use the reliability-filtered POS dataset
# ------------------------------------------------

h9_test = pos_summary_filtered.copy()

# ------------------------------------------------
# 2. Build POS × recall-outcome contingency table
# ------------------------------------------------

contingency = h9_test.set_index("pos")[
    ["total_correct", "total_incorrect"]
]

print("\nOBSERVED RECALL OUTCOMES BY POS")
print("-" * 80)

display(contingency)

# ------------------------------------------------
# 3. Chi-square test
# ------------------------------------------------

chi2, p_value, dof, expected = chi2_contingency(
    contingency
)

# ------------------------------------------------
# 4. Cramer's V
# ------------------------------------------------

n = contingency.to_numpy().sum()

min_dimension = min(
    contingency.shape[0] - 1,
    contingency.shape[1] - 1
)

cramers_v = np.sqrt(
    chi2 / (n * min_dimension)
)

# ------------------------------------------------
# 5. Print results
# ------------------------------------------------

print("\n" + "=" * 80)
print("CHI-SQUARE TEST")
print("=" * 80)

print(f"Chi-square statistic : {chi2:.2f}")
print(f"Degrees of freedom  : {dof}")
print(f"P-value             : {p_value:.10f}")

print("\n" + "=" * 80)
print("EFFECT SIZE")
print("=" * 80)

print(f"Cramer's V           : {cramers_v:.6f}")

# ------------------------------------------------
# 6. Hypothesis decision
# ------------------------------------------------

alpha = 0.05

print("\n" + "=" * 80)
print("HYPOTHESES")
print("=" * 80)

print("H0: Word Recall Rate is the same across POS categories")
print("H1: Word Recall Rate differs across POS categories")

print("\nDecision rule")
print("-" * 80)

print(f"Significance level (α): {alpha}")

if p_value < alpha:
    print("Result: Reject H0")
    print(
        "Evidence indicates that recall differs significantly "
        "across POS categories."
    )
else:
    print("Result: Fail to reject H0")
    print(
        "There is insufficient evidence that recall differs "
        "across POS categories."
    )

# ------------------------------------------------
# 7. POS ranking
# ------------------------------------------------

pos_ranking = h9_test[
    [
        "pos",
        "n_words",
        "n_traces",
        "total_attempts",
        "recall_%",
        "incorrect_%"
    ]
].copy()

pos_ranking = pos_ranking.sort_values(
    "recall_%",
    ascending=False
).reset_index(drop=True)

pos_ranking["recall_rank"] = (
    pos_ranking.index + 1
)

print("\n" + "=" * 80)
print("POS RECALL RANKING")
print("=" * 80)

display(pos_ranking)

# ------------------------------------------------
# 8. Interpretation guide
# ------------------------------------------------

print("\n" + "=" * 80)
print("INTERPRETATION")
print("=" * 80)

print(
    "Chi-square tests whether POS category and recall outcome "
    "are statistically associated."
)

print(
    "Cramer's V measures the magnitude of the association."
)

print(
    "A statistically significant result does not necessarily "
    "mean the practical difference is large."
)

print("\n" + "=" * 80)
print("H9 PHASE 2 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H9 — PART OF SPEECH → RECALL
# PHASE 3: CONTROLLED POS EFFECT
# ================================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf

print("=" * 80)
print("H9 — PART OF SPEECH → RECALL")
print("PHASE 3: CONTROLLED POS EFFECT")
print("=" * 80)


# ------------------------------------------------
# 1. Create analysis dataset
# ------------------------------------------------

h9_data = traces_cleaned.copy()


# ------------------------------------------------
# 2. Keep only valid recall observations
# ------------------------------------------------

h9_data = h9_data[
    h9_data["pos"].notna() &
    h9_data["learning_language"].notna() &
    h9_data["lag_days"].notna() &
    h9_data["history_seen"].notna() &
    h9_data["session_seen"].notna() &
    h9_data["session_correct"].notna()
].copy()


# ------------------------------------------------
# 3. Keep POS categories with at least 20 words
#    This matches H9 Phase 1 reliability rule
# ------------------------------------------------

pos_word_counts = (
    h9_data
    .groupby("pos")["surface_form"]
    .nunique()
)

valid_pos = pos_word_counts[
    pos_word_counts >= 20
].index

h9_data = h9_data[
    h9_data["pos"].isin(valid_pos)
].copy()


# ------------------------------------------------
# 4. Create transformed control variables
# ------------------------------------------------

h9_data["log_lag"] = np.log1p(
    h9_data["lag_days"].clip(lower=0)
)

h9_data["log_exposure"] = np.log1p(
    h9_data["history_seen"].clip(lower=0)
)


# ------------------------------------------------
# 5. Create recall proportion
# ------------------------------------------------

h9_data["recall_rate"] = (
    h9_data["session_correct"] /
    h9_data["session_seen"]
)

# Remove impossible values
h9_data = h9_data[
    np.isfinite(h9_data["recall_rate"]) &
    (h9_data["recall_rate"] >= 0) &
    (h9_data["recall_rate"] <= 1) &
    (h9_data["session_seen"] > 0)
].copy()


# ------------------------------------------------
# 6. Make n the reference POS
# ------------------------------------------------

if "n" not in h9_data["pos"].unique():
    raise ValueError(
        "Reference POS 'n' is not available in the analysis data."
    )

h9_data["pos"] = pd.Categorical(
    h9_data["pos"],
    categories=["n"] + [
        x for x in sorted(h9_data["pos"].unique())
        if x != "n"
    ]
)


# ------------------------------------------------
# 7. Print model data information
# ------------------------------------------------

print("\nMODEL DATA")
print("-" * 80)

print(f"Observations used       : {len(h9_data):,}")
print(
    f"Total recall attempts   : "
    f"{h9_data['session_seen'].sum():,}"
)
print(
    f"POS categories          : "
    f"{h9_data['pos'].nunique()}"
)

print(
    f"POS categories tested   : "
    f"{list(h9_data['pos'].cat.categories)}"
)


# ------------------------------------------------
# 8. Fit controlled logistic model
# ------------------------------------------------
#
# Because each row can represent multiple recall attempts,
# session_seen is used as the frequency weight.
#
# Outcome:
#     session_correct / session_seen
#
# Predictors:
#     POS
#     log_lag
#     log_exposure
#     learning_language
# ------------------------------------------------

formula = """
recall_rate ~
C(pos, Treatment(reference='n'))
+ log_lag
+ log_exposure
+ C(learning_language)
"""

model = smf.glm(
    formula=formula,
    data=h9_data,
    family=sm.families.Binomial(),
    freq_weights=h9_data["session_seen"]
).fit()


# ------------------------------------------------
# 9. Extract POS effects
# ------------------------------------------------

params = model.params
pvalues = model.pvalues
conf_int = model.conf_int()

pos_terms = [
    term for term in params.index
    if "C(pos" in term
]

pos_results = pd.DataFrame({
    "term": pos_terms,
    "coefficient": [
        params[t] for t in pos_terms
    ],
    "odds_ratio": [
        np.exp(params[t]) for t in pos_terms
    ],
    "p_value": [
        pvalues[t] for t in pos_terms
    ],
    "ci_lower": [
        np.exp(conf_int.loc[t, 0])
        for t in pos_terms
    ],
    "ci_upper": [
        np.exp(conf_int.loc[t, 1])
        for t in pos_terms
    ]
})

pos_results = pos_results.sort_values(
    "odds_ratio",
    ascending=False
).reset_index(drop=True)


# ------------------------------------------------
# 10. Display POS effects
# ------------------------------------------------

print("\n" + "=" * 80)
print("CONTROLLED POS EFFECTS")
print("=" * 80)

print("\nReference POS: n")

display(pos_results)


# ------------------------------------------------
# 11. Control-variable effects
# ------------------------------------------------

print("\n" + "=" * 80)
print("CONTROL VARIABLES")
print("=" * 80)

for variable in ["log_lag", "log_exposure"]:

    print(
        f"{variable:<20} "
        f"coefficient = {params[variable]:.6f}   "
        f"p-value = {pvalues[variable]:.10f}"
    )


# ------------------------------------------------
# 12. Model fit
# ------------------------------------------------

print("\n" + "=" * 80)
print("MODEL FIT")
print("=" * 80)

print(f"Deviance            : {model.deviance:.2f}")
print(f"Null deviance       : {model.null_deviance:.2f}")
print(f"AIC                 : {model.aic:.2f}")


# ------------------------------------------------
# 13. Overall POS significance
# ------------------------------------------------
#
# Compare:
#   Model 0 = controls only
#   Model 1 = controls + POS
#
# This is more appropriate than relying only on
# individual POS p-values.
# ------------------------------------------------

control_formula = """
recall_rate ~
log_lag
+ log_exposure
+ C(learning_language)
"""

control_model = smf.glm(
    formula=control_formula,
    data=h9_data,
    family=sm.families.Binomial(),
    freq_weights=h9_data["session_seen"]
).fit()


# Likelihood-ratio test
lr_stat = (
    control_model.deviance -
    model.deviance
)

df_diff = (
    control_model.df_resid -
    model.df_resid
)

from scipy.stats import chi2

lr_pvalue = chi2.sf(
    lr_stat,
    df_diff
)


# ------------------------------------------------
# 14. Overall POS hypothesis test
# ------------------------------------------------

print("\n" + "=" * 80)
print("OVERALL POS SIGNIFICANCE TEST")
print("=" * 80)

print(
    f"Likelihood-ratio statistic : {lr_stat:.2f}"
)

print(
    f"Degrees of freedom         : {int(df_diff)}"
)

print(
    f"P-value                    : {lr_pvalue:.10f}"
)


# ------------------------------------------------
# 15. Hypothesis decision
# ------------------------------------------------

alpha = 0.05

print("\n" + "=" * 80)
print("H9 HYPOTHESIS DECISION")
print("=" * 80)

print(
    "H0: POS has no independent effect on recall "
    "after controlling for lag, prior exposure, "
    "and learning language"
)

print(
    "H1: POS has an independent effect on recall "
    "after controlling for lag, prior exposure, "
    "and learning language"
)

print("\nDecision")
print("-" * 80)

if lr_pvalue < alpha:
    print("Reject H0")
    print(
        "Evidence suggests POS contributes independently "
        "to recall differences after controlling for "
        "lag, prior exposure, and learning language."
    )
else:
    print("Fail to reject H0")
    print(
        "There is insufficient evidence that POS contributes "
        "an independent effect on recall after controlling "
        "for lag, prior exposure, and learning language."
    )


# ------------------------------------------------
# 16. Interpretation guide
# ------------------------------------------------

print("\n" + "=" * 80)
print("INTERPRETATION GUIDE")
print("=" * 80)

print(
    "Odds Ratio > 1 → higher odds of correct recall than POS 'n'."
)

print(
    "Odds Ratio < 1 → lower odds of correct recall than POS 'n'."
)

print(
    "95% CI excluding 1 indicates a statistically significant "
    "difference from the reference POS at α = 0.05."
)

print(
    "The overall likelihood-ratio test determines whether POS "
    "as a group adds explanatory value beyond the control variables."
)

print(
    "These results represent associations, not causal effects."
)


print("\n" + "=" * 80)
print("H9 PHASE 3 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H10 — POS EFFECT DIFFERS BY LEARNING LANGUAGE
# PHASE 1: DESCRIPTIVE INTERACTION ANALYSIS
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 80)
print("H10 — POS EFFECT DIFFERS BY LEARNING LANGUAGE")
print("PHASE 1: DESCRIPTIVE INTERACTION ANALYSIS")
print("=" * 80)


# ------------------------------------------------
# 1. Start from the cleaned trace-level dataset
# ------------------------------------------------

h10_data = traces_cleaned.copy()


# ------------------------------------------------
# 2. Basic validity filtering
# ------------------------------------------------

h10_data = h10_data[
    h10_data["learning_language"].notna() &
    h10_data["pos"].notna() &
    h10_data["surface_form"].notna() &
    h10_data["session_seen"].notna() &
    h10_data["session_correct"].notna()
].copy()

h10_data = h10_data[
    h10_data["session_seen"] > 0
].copy()


# ------------------------------------------------
# 3. Calculate unique words per
#    learning language × POS combination
# ------------------------------------------------

word_counts = (
    h10_data
    .groupby(
        ["learning_language", "pos"]
    )["surface_form"]
    .nunique()
    .reset_index(name="n_words")
)


# ------------------------------------------------
# 4. Reliability filter
#
# At least 20 unique words per
# language × POS combination
# ------------------------------------------------

MIN_WORDS = 20

valid_combinations = word_counts[
    word_counts["n_words"] >= MIN_WORDS
][
    ["learning_language", "pos"]
]

h10_data = h10_data.merge(
    valid_combinations,
    on=["learning_language", "pos"],
    how="inner"
)


# ------------------------------------------------
# 5. Aggregate recall
# ------------------------------------------------

h10_summary = (
    h10_data
    .groupby(
        ["learning_language", "pos"]
    )
    .agg(
        n_words=("surface_form", "nunique"),
        n_traces=("surface_form", "size"),
        total_attempts=("session_seen", "sum"),
        total_correct=("session_correct", "sum")
    )
    .reset_index()
)


h10_summary["total_incorrect"] = (
    h10_summary["total_attempts"] -
    h10_summary["total_correct"]
)

h10_summary["recall_%"] = (
    h10_summary["total_correct"] /
    h10_summary["total_attempts"] *
    100
)

h10_summary["incorrect_%"] = (
    h10_summary["total_incorrect"] /
    h10_summary["total_attempts"] *
    100
)


# ------------------------------------------------
# 6. Print model/descriptive data information
# ------------------------------------------------

print("\nMODEL / DESCRIPTIVE DATA")
print("-" * 80)

print(
    f"Observations used        : {len(h10_data):,}"
)

print(
    f"Learning languages       : "
    f"{h10_summary['learning_language'].nunique()}"
)

print(
    f"POS categories           : "
    f"{h10_summary['pos'].nunique()}"
)

print(
    f"Language × POS groups    : "
    f"{len(h10_summary):,}"
)

print(
    f"Minimum words per group  : "
    f"{MIN_WORDS}"
)


# ------------------------------------------------
# 7. Full language × POS summary
# ------------------------------------------------

print("\n" + "=" * 80)
print("LANGUAGE × POS RECALL SUMMARY")
print("=" * 80)

h10_display = h10_summary[
    [
        "learning_language",
        "pos",
        "n_words",
        "n_traces",
        "total_attempts",
        "total_correct",
        "recall_%",
        "incorrect_%"
    ]
].sort_values(
    ["learning_language", "recall_%"],
    ascending=[True, False]
).reset_index(drop=True)

display(h10_display)


# ------------------------------------------------
# 8. Create recall heatmap
# ------------------------------------------------

heatmap_data = (
    h10_summary
    .pivot(
        index="pos",
        columns="learning_language",
        values="recall_%"
    )
)

plt.figure(figsize=(10, 9))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5,
    cbar_kws={"label": "Recall Rate (%)"}
)

plt.title(
    "Word Recall Rate by POS and Learning Language"
)

plt.xlabel("Learning Language")
plt.ylabel("Part of Speech")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 9. Within-language POS spread
#
# This tells us whether POS differences are
# particularly large in some languages.
# ------------------------------------------------

language_pos_spread = (
    h10_summary
    .groupby("learning_language")
    .agg(
        n_pos=("pos", "nunique"),
        highest_recall_pct=("recall_%", "max"),
        lowest_recall_pct=("recall_%", "min")
    )
    .reset_index()
)

language_pos_spread["pos_recall_spread_pp"] = (
    language_pos_spread["highest_recall_pct"] -
    language_pos_spread["lowest_recall_pct"]
)

language_pos_spread = language_pos_spread.sort_values(
    "pos_recall_spread_pp",
    ascending=False
)


print("\n" + "=" * 80)
print("WITHIN-LANGUAGE POS RECALL SPREAD")
print("=" * 80)

display(language_pos_spread)


# ------------------------------------------------
# 10. Identify highest and lowest POS
#     within every language
# ------------------------------------------------

highest_pos = (
    h10_summary
    .loc[
        h10_summary
        .groupby("learning_language")["recall_%"]
        .idxmax()
    ]
    [
        [
            "learning_language",
            "pos",
            "recall_%",
            "n_words",
            "total_attempts"
        ]
    ]
    .sort_values("learning_language")
)

lowest_pos = (
    h10_summary
    .loc[
        h10_summary
        .groupby("learning_language")["recall_%"]
        .idxmin()
    ]
    [
        [
            "learning_language",
            "pos",
            "recall_%",
            "n_words",
            "total_attempts"
        ]
    ]
    .sort_values("learning_language")
)


print("\n" + "=" * 80)
print("HIGHEST-RECALL POS WITHIN EACH LANGUAGE")
print("=" * 80)

display(highest_pos)


print("\n" + "=" * 80)
print("LOWEST-RECALL POS WITHIN EACH LANGUAGE")
print("=" * 80)

display(lowest_pos)


# ------------------------------------------------
# 11. POS consistency across languages
#
# For each POS, calculate the range of recall
# across languages.
# ------------------------------------------------

pos_cross_language = (
    h10_summary
    .groupby("pos")
    .agg(
        n_languages=("learning_language", "nunique"),
        avg_recall_pct=("recall_%", "mean"),
        min_recall_pct=("recall_%", "min"),
        max_recall_pct=("recall_%", "max")
    )
    .reset_index()
)

pos_cross_language["language_recall_range_pp"] = (
    pos_cross_language["max_recall_pct"] -
    pos_cross_language["min_recall_pct"]
)

pos_cross_language = pos_cross_language.sort_values(
    "language_recall_range_pp",
    ascending=False
)


print("\n" + "=" * 80)
print("POS VARIATION ACROSS LEARNING LANGUAGES")
print("=" * 80)

display(pos_cross_language)


# ------------------------------------------------
# 12. Key descriptive findings
# ------------------------------------------------

overall_min = h10_summary.loc[
    h10_summary["recall_%"].idxmin()
]

overall_max = h10_summary.loc[
    h10_summary["recall_%"].idxmax()
]

largest_language_spread = language_pos_spread.iloc[0]

largest_pos_variation = pos_cross_language.iloc[0]


print("\n" + "=" * 80)
print("H10 PHASE 1 KEY FINDINGS")
print("=" * 80)

print(
    f"Lowest language × POS recall : "
    f"{overall_min['learning_language']} × "
    f"{overall_min['pos']} → "
    f"{overall_min['recall_%']:.2f}%"
)

print(
    f"Highest language × POS recall: "
    f"{overall_max['learning_language']} × "
    f"{overall_max['pos']} → "
    f"{overall_max['recall_%']:.2f}%"
)

print(
    f"Largest within-language POS spread: "
    f"{largest_language_spread['learning_language']} → "
    f"{largest_language_spread['pos_recall_spread_pp']:.2f} pp"
)

print(
    f"Largest cross-language variation for one POS: "
    f"{largest_pos_variation['pos']} → "
    f"{largest_pos_variation['language_recall_range_pp']:.2f} pp"
)


print("\n" + "=" * 80)
print("H10 PHASE 1 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H10 — POS EFFECT DIFFERS BY LEARNING LANGUAGE
# PHASE 2: FORMAL POS × LANGUAGE INTERACTION TEST
# ================================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import chi2

print("=" * 80)
print("H10 — POS EFFECT DIFFERS BY LEARNING LANGUAGE")
print("PHASE 2: FORMAL POS × LANGUAGE INTERACTION TEST")
print("=" * 80)


# ------------------------------------------------
# 1. Prepare data
# ------------------------------------------------

h10_model = h10_data.copy()


# ------------------------------------------------
# 2. Create transformed control variables
# ------------------------------------------------

h10_model["log_lag"] = np.log1p(
    h10_model["lag_days"].clip(lower=0)
)

h10_model["log_exposure"] = np.log1p(
    h10_model["history_seen"].clip(lower=0)
)


# ------------------------------------------------
# 3. Create recall proportion
# ------------------------------------------------

h10_model["recall_rate"] = (
    h10_model["session_correct"] /
    h10_model["session_seen"]
)


# ------------------------------------------------
# 4. Remove invalid observations
# ------------------------------------------------

h10_model = h10_model[
    h10_model["session_seen"] > 0
].copy()

h10_model = h10_model[
    h10_model["recall_rate"].between(0, 1)
].copy()

h10_model = h10_model[
    h10_model["log_lag"].notna() &
    h10_model["log_exposure"].notna()
].copy()


print("\nMODEL DATA")
print("-" * 80)

print(
    f"Observations used       : {len(h10_model):,}"
)

print(
    f"Total recall attempts   : "
    f"{h10_model['session_seen'].sum():,.0f}"
)

print(
    f"Learning languages      : "
    f"{h10_model['learning_language'].nunique()}"
)

print(
    f"POS categories          : "
    f"{h10_model['pos'].nunique()}"
)


# ------------------------------------------------
# 5. Define formulas
# ------------------------------------------------
#
# Model 1:
# POS + language + lag + exposure
#
# Model 2:
# POS + language + POS × language
# + lag + exposure
#
# The interaction tests whether POS effects
# change depending on learning language.
# ------------------------------------------------

formula_reduced = """
recall_rate ~
C(pos) +
C(learning_language) +
log_lag +
log_exposure
"""

formula_full = """
recall_rate ~
C(pos) * C(learning_language) +
log_lag +
log_exposure
"""


# ------------------------------------------------
# 6. Fit reduced model
# ------------------------------------------------

print("\n" + "=" * 80)
print("MODEL 1 — MAIN EFFECTS ONLY")
print("=" * 80)

model_reduced = smf.glm(
    formula=formula_reduced,
    data=h10_model,
    family=sm.families.Binomial(),
    freq_weights=h10_model["session_seen"]
).fit()

print(
    f"Deviance : {model_reduced.deviance:.2f}"
)

print(
    f"AIC      : {model_reduced.aic:.2f}"
)


# ------------------------------------------------
# 7. Fit full interaction model
# ------------------------------------------------

print("\n" + "=" * 80)
print("MODEL 2 — POS × LEARNING LANGUAGE INTERACTION")
print("=" * 80)

model_full = smf.glm(
    formula=formula_full,
    data=h10_model,
    family=sm.families.Binomial(),
    freq_weights=h10_model["session_seen"]
).fit()

print(
    f"Deviance : {model_full.deviance:.2f}"
)

print(
    f"AIC      : {model_full.aic:.2f}"
)


# ------------------------------------------------
# 8. Likelihood-ratio / deviance comparison
# ------------------------------------------------

lr_stat = (
    model_reduced.deviance -
    model_full.deviance
)

df_diff = (
    model_full.df_model -
    model_reduced.df_model
)

p_value = chi2.sf(
    lr_stat,
    df_diff
)


print("\n" + "=" * 80)
print("POS × LANGUAGE INTERACTION TEST")
print("=" * 80)

print(
    f"Likelihood-ratio statistic : {lr_stat:.4f}"
)

print(
    f"Degrees of freedom         : {int(df_diff)}"
)

print(
    f"P-value                    : {p_value:.10f}"
)


# ------------------------------------------------
# 9. Hypothesis decision
# ------------------------------------------------

alpha = 0.05

print("\n" + "=" * 80)
print("HYPOTHESES")
print("=" * 80)

print(
    "H0: The effect of POS is the same across learning languages."
)

print(
    "H1: The effect of POS differs across learning languages."
)

print("\nDecision")
print("-" * 80)

if p_value < alpha:

    print(
        "Reject H0"
    )

    print(
        "Evidence suggests that POS effects differ "
        "across learning languages."
    )

else:

    print(
        "Fail to reject H0"
    )

    print(
        "There is insufficient evidence that POS effects "
        "differ across learning languages."
    )


# ------------------------------------------------
# 10. Extract interaction coefficients
# ------------------------------------------------

interaction_terms = [
    term for term in model_full.params.index
    if ":" in term
]

interaction_results = pd.DataFrame({
    "term": interaction_terms,
    "coefficient": [
        model_full.params[t]
        for t in interaction_terms
    ],
    "p_value": [
        model_full.pvalues[t]
        for t in interaction_terms
    ]
})

interaction_results["odds_ratio"] = np.exp(
    interaction_results["coefficient"]
)

interaction_results["ci_lower"] = np.exp(
    [
        model_full.conf_int().loc[t, 0]
        for t in interaction_terms
    ]
)

interaction_results["ci_upper"] = np.exp(
    [
        model_full.conf_int().loc[t, 1]
        for t in interaction_terms
    ]
)

interaction_results = interaction_results.sort_values(
    "p_value"
).reset_index(drop=True)


# ------------------------------------------------
# 11. Display interaction effects
# ------------------------------------------------

print("\n" + "=" * 80)
print("POS × LANGUAGE INTERACTION EFFECTS")
print("=" * 80)

display(interaction_results)


# ------------------------------------------------
# 12. Control-variable results
# ------------------------------------------------

print("\n" + "=" * 80)
print("CONTROL VARIABLES")
print("=" * 80)

control_terms = [
    "log_lag",
    "log_exposure"
]

control_results = pd.DataFrame({
    "term": control_terms,
    "coefficient": [
        model_full.params[t]
        for t in control_terms
    ],
    "p_value": [
        model_full.pvalues[t]
        for t in control_terms
    ]
})

display(control_results)


# ------------------------------------------------
# 13. Final summary
# ------------------------------------------------

print("\n" + "=" * 80)
print("H10 PHASE 2 SUMMARY")
print("=" * 80)

print(
    f"Reduced-model deviance : "
    f"{model_reduced.deviance:.2f}"
)

print(
    f"Full-model deviance    : "
    f"{model_full.deviance:.2f}"
)

print(
    f"Interaction LR statistic: "
    f"{lr_stat:.2f}"
)

print(
    f"Interaction degrees of freedom: "
    f"{int(df_diff)}"
)

print(
    f"Interaction p-value: "
    f"{p_value:.10f}"
)

if p_value < alpha:
    print(
        "Conclusion: POS × learning language interaction "
        "is statistically significant."
    )
else:
    print(
        "Conclusion: POS × learning language interaction "
        "is not statistically significant."
    )

print("\n" + "=" * 80)
print("H10 PHASE 2 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# H10 — POS EFFECT DIFFERS BY LEARNING LANGUAGE
# PHASE 3: PRACTICAL POS × LANGUAGE EFFECTS
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 80)
print("H10 — POS EFFECT DIFFERS BY LEARNING LANGUAGE")
print("PHASE 3: PRACTICAL POS × LANGUAGE EFFECTS")
print("=" * 80)


# ------------------------------------------------
# 1. Use the same model data from Phase 2
# ------------------------------------------------

phase3_data = h10_model.copy()


# ------------------------------------------------
# 2. Representative control values
# ------------------------------------------------
#
# Use medians so that predictions are not driven
# by extreme lag or exposure observations.
# ------------------------------------------------

median_lag = phase3_data["log_lag"].median()
median_exposure = phase3_data["log_exposure"].median()

median_lag_days = phase3_data["lag_days"].median()
median_exposure_raw = phase3_data["history_seen"].median()


print("\nCONTROL VALUES")
print("-" * 80)

print(
    f"Median lag days       : {median_lag_days:.2f}"
)

print(
    f"Median prior exposure : {median_exposure_raw:.2f}"
)

print(
    f"Median log lag        : {median_lag:.4f}"
)

print(
    f"Median log exposure   : {median_exposure:.4f}"
)


# ------------------------------------------------
# 3. Identify observed language × POS combinations
# ------------------------------------------------

observed_groups = (
    phase3_data[
        [
            "learning_language",
            "pos"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        ["learning_language", "pos"]
    )
    .reset_index(drop=True)
)


print("\nOBSERVED LANGUAGE × POS GROUPS")
print("-" * 80)

print(
    f"Observed language × POS groups : "
    f"{len(observed_groups)}"
)


# ------------------------------------------------
# 4. Build prediction dataset
# ------------------------------------------------

prediction_data = observed_groups.copy()

prediction_data["log_lag"] = median_lag
prediction_data["log_exposure"] = median_exposure


# ------------------------------------------------
# 5. Predict adjusted recall
# ------------------------------------------------
#
# We use the full interaction model from Phase 2.
# ------------------------------------------------

prediction_data["adjusted_recall"] = (
    model_full.predict(prediction_data)
)

prediction_data["adjusted_recall_%"] = (
    prediction_data["adjusted_recall"] * 100
)


# ------------------------------------------------
# 6. Display adjusted recall matrix
# ------------------------------------------------

adjusted_matrix = prediction_data.pivot(
    index="pos",
    columns="learning_language",
    values="adjusted_recall_%"
)

print("\n" + "=" * 80)
print("MODEL-ADJUSTED RECALL BY POS × LEARNING LANGUAGE")
print("=" * 80)

display(
    adjusted_matrix.round(2)
)


# ------------------------------------------------
# 7. Within-language POS spread
# ------------------------------------------------

within_language = (
    prediction_data
    .groupby("learning_language")
    .agg(
        n_pos=("pos", "nunique"),
        highest_recall_pct=(
            "adjusted_recall_%",
            "max"
        ),
        lowest_recall_pct=(
            "adjusted_recall_%",
            "min"
        )
    )
    .reset_index()
)

within_language["pos_recall_spread_pp"] = (
    within_language["highest_recall_pct"]
    -
    within_language["lowest_recall_pct"]
)

within_language = within_language.sort_values(
    "pos_recall_spread_pp",
    ascending=False
)


print("\n" + "=" * 80)
print("WITHIN-LANGUAGE POS RECALL SPREAD")
print("=" * 80)

display(
    within_language.round(2)
)


# ------------------------------------------------
# 8. Cross-language variation by POS
# ------------------------------------------------

pos_variation = (
    prediction_data
    .groupby("pos")
    .agg(
        n_languages=("learning_language", "nunique"),
        avg_recall_pct=(
            "adjusted_recall_%",
            "mean"
        ),
        min_recall_pct=(
            "adjusted_recall_%",
            "min"
        ),
        max_recall_pct=(
            "adjusted_recall_%",
            "max"
        )
    )
    .reset_index()
)

pos_variation["language_recall_range_pp"] = (
    pos_variation["max_recall_pct"]
    -
    pos_variation["min_recall_pct"]
)

pos_variation = pos_variation.sort_values(
    "language_recall_range_pp",
    ascending=False
)


print("\n" + "=" * 80)
print("POS VARIATION ACROSS LEARNING LANGUAGES")
print("=" * 80)

display(
    pos_variation.round(2)
)


# ------------------------------------------------
# 9. Find highest and lowest adjusted
#    language × POS combinations
# ------------------------------------------------

lowest_combo = (
    prediction_data
    .sort_values("adjusted_recall_%")
    .iloc[0]
)

highest_combo = (
    prediction_data
    .sort_values("adjusted_recall_%")
    .iloc[-1]
)


print("\n" + "=" * 80)
print("EXTREME ADJUSTED RECALL COMBINATIONS")
print("=" * 80)

print(
    f"Lowest adjusted recall  : "
    f"{lowest_combo['learning_language']} × "
    f"{lowest_combo['pos']} → "
    f"{lowest_combo['adjusted_recall_%']:.2f}%"
)

print(
    f"Highest adjusted recall : "
    f"{highest_combo['learning_language']} × "
    f"{highest_combo['pos']} → "
    f"{highest_combo['adjusted_recall_%']:.2f}%"
)

print(
    f"Overall adjusted spread : "
    f"{highest_combo['adjusted_recall_%'] - lowest_combo['adjusted_recall_%']:.2f} pp"
)


# ------------------------------------------------
# 10. Heatmap
# ------------------------------------------------

plt.figure(figsize=(11, 8))

sns.heatmap(
    adjusted_matrix,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5,
    cbar_kws={
        "label": "Model-Adjusted Recall (%)"
    }
)

plt.title(
    "Model-Adjusted Word Recall by POS and Learning Language\n"
    f"(Lag = {median_lag_days:.1f} days, "
    f"Prior Exposure = {median_exposure_raw:.0f})"
)

plt.xlabel("Learning Language")
plt.ylabel("Part of Speech")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 11. Final H10 interpretation
# ------------------------------------------------

print("\n" + "=" * 80)
print("H10 PHASE 3 BUSINESS INTERPRETATION")
print("=" * 80)

print(
    "The POS × learning-language interaction is statistically significant."
)

print(
    "This means POS effects are not uniform across learning languages."
)

print(
    "The adjusted recall analysis identifies where the largest "
    "practical POS differences occur while holding lag and prior "
    "exposure constant."
)

print(
    "\nImportant: these are observational model-adjusted associations, "
    "not causal effects."
)

print("\n" + "=" * 80)
print("H10 PHASE 3 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# H1 — STEP 1: LAG vs RECALL — DISTRIBUTION & CORRELATION
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Load data ────────────────────────────────────────────────
traces_df = pd.read_csv("traces_cleaned.csv")

print("=" * 60)
print("H1 — STEP 1: LAG vs RECALL")
print("=" * 60)

# ── 1. Basic shape check ─────────────────────────────────────
print(f"\nTotal records : {len(traces_df):,}")
print(f"lag_days range: {traces_df['lag_days'].min():.2f} → {traces_df['lag_days'].max():.2f}")
print(f"p_recall range: {traces_df['p_recall'].min():.4f} → {traces_df['p_recall'].max():.4f}")
print(f"p_recall mean : {traces_df['p_recall'].mean():.4f}")

# ── 2. lag_days percentile distribution ──────────────────────
print("\n── lag_days percentile distribution ──")
percentiles = [0, 10, 25, 50, 75, 90, 95, 99, 100]
for p in percentiles:
    print(f"  P{p:>3}  : {np.percentile(traces_df['lag_days'], p):.2f} days")

# ── 3. Cap outliers for correlation (use P99 cap) ────────────
lag_cap = np.percentile(traces_df['lag_days'], 99)
df_capped = traces_df[traces_df['lag_days'] <= lag_cap].copy()
print(f"\nRecords after P99 lag cap ({lag_cap:.1f} days): {len(df_capped):,}")

# ── 4. Spearman correlation (full + capped) ──────────────────
print("\n── Spearman Correlation: lag_days vs p_recall ──")

rho_full, p_full = stats.spearmanr(
    traces_df['lag_days'], traces_df['p_recall']
)
rho_capped, p_capped = stats.spearmanr(
    df_capped['lag_days'], df_capped['p_recall']
)

print(f"  Full dataset  : rho = {rho_full:.4f}, p-value = {p_full:.2e}")
print(f"  P99 capped    : rho = {rho_capped:.4f}, p-value = {p_capped:.2e}")

# ── 5. Interpretation ────────────────────────────────────────
alpha = 0.05
print("\n── H1 Step 1 Decision ──")
if p_capped < alpha and rho_capped < 0:
    print("  Result   : SUPPORTS H1")
    print(f"  Recall decreases as lag increases (rho = {rho_capped:.4f})")
elif p_capped < alpha and rho_capped > 0:
    print("  Result   : CONTRADICTS H1")
    print(f"  Recall increases with lag — unexpected (rho = {rho_capped:.4f})")
else:
    print("  Result   : INSUFFICIENT EVIDENCE to reject H0")

# ── 6. Mean recall by lag_days bucket (decile bins) ──────────
print("\n── Mean Recall by Lag Decile ──")
df_capped['lag_decile'] = pd.qcut(
    df_capped['lag_days'], q=10, duplicates='drop'
)
decile_recall = (
    df_capped.groupby('lag_decile', observed=True)['p_recall']
    .agg(['mean', 'count'])
    .reset_index()
)
decile_recall.columns = ['lag_decile', 'avg_recall', 'n_records']
decile_recall['avg_recall_%'] = (decile_recall['avg_recall'] * 100).round(2)
print(decile_recall[['lag_decile', 'avg_recall_%', 'n_records']].to_string(index=False))
```

```python
# ============================================================
# H1 — STEP 2: LAG BIN BREAKDOWN + KRUSKAL-WALLIS TEST
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Load data ────────────────────────────────────────────────
traces_df = pd.read_csv("traces_cleaned.csv")
forgetting_curve_df = pd.read_csv("forgetting_curve_cleaned.csv")

print("=" * 60)
print("H1 — STEP 2: LAG BIN BREAKDOWN + STATISTICAL TEST")
print("=" * 60)

# ── 1. Create lag bins matching Duolingo's forgetting curve ──
lag_bins    = [0, 1/24, 6/24, 1, 3, 7, 14, 28, 91, 9999]
lag_labels  = [
    '<1 hour', '1-6 hours', '6-24 hours',
    '1-3 days', '3-7 days', '1-2 weeks',
    '2-4 weeks', '1-3 months', '3+ months'
]
lag_order = {label: i for i, label in enumerate(lag_labels)}

traces_df['lag_bin'] = pd.cut(
    traces_df['lag_days'],
    bins=lag_bins,
    labels=lag_labels,
    right=True
)

# ── 2. Recall summary by lag bin ─────────────────────────────
print("\n── Mean Recall by Lag Bin ──")
bin_summary = (
    traces_df.groupby('lag_bin', observed=True)['p_recall']
    .agg(['mean', 'median', 'std', 'count'])
    .reset_index()
)
bin_summary.columns = ['lag_bin', 'avg_recall', 'median_recall', 'std_recall', 'n_records']
bin_summary['avg_recall_%']    = (bin_summary['avg_recall'] * 100).round(2)
bin_summary['median_recall_%'] = (bin_summary['median_recall'] * 100).round(2)
bin_summary['lag_order']       = bin_summary['lag_bin'].map(lag_order)
bin_summary = bin_summary.sort_values('lag_order')

print(bin_summary[
    ['lag_bin', 'avg_recall_%', 'median_recall_%', 'std_recall', 'n_records']
].to_string(index=False))

# ── 3. Recall drop from best to worst bin ────────────────────
best_recall  = bin_summary['avg_recall_%'].max()
worst_recall = bin_summary['avg_recall_%'].min()
print(f"\nRecall drop from best to worst lag bin: "
      f"{best_recall:.2f}% → {worst_recall:.2f}% "
      f"({best_recall - worst_recall:.2f} percentage points)")

# ── 4. Kruskal-Wallis test across all lag bins ───────────────
print("\n── Kruskal-Wallis Test (are all lag bins the same?) ──")
groups = [
    traces_df[traces_df['lag_bin'] == lb]['p_recall'].dropna().values
    for lb in lag_labels
    if lb in traces_df['lag_bin'].cat.categories
]
kw_stat, kw_p = stats.kruskal(*groups)
print(f"  H-statistic : {kw_stat:.2f}")
print(f"  p-value     : {kw_p:.2e}")

alpha = 0.05
if kw_p < alpha:
    print("  Decision    : REJECT H0 — recall differs significantly across lag bins")
else:
    print("  Decision    : FAIL TO REJECT H0")

# ── 5. Pairwise comparison — adjacent bins ───────────────────
print("\n── Adjacent Bin Pairwise Mann-Whitney U Tests ──")
print(f"{'Comparison':<35} {'U-stat':>10} {'p-value':>12} {'Significant':>12}")
print("-" * 72)

for i in range(len(lag_labels) - 1):
    bin_a = lag_labels[i]
    bin_b = lag_labels[i + 1]
    group_a = traces_df[traces_df['lag_bin'] == bin_a]['p_recall'].dropna().values
    group_b = traces_df[traces_df['lag_bin'] == bin_b]['p_recall'].dropna().values
    if len(group_a) > 0 and len(group_b) > 0:
        u_stat, u_p = stats.mannwhitneyu(group_a, group_b, alternative='greater')
        sig = "✓ Yes" if u_p < alpha else "✗ No"
        print(f"  {bin_a:<15} > {bin_b:<15} {u_stat:>12,.0f} {u_p:>12.2e} {sig:>10}")

# ── 6. Forgetting curve cross-check ─────────────────────────
print("\n── Cross-check: Forgetting Curve Dataset (All Languages) ──")
fc_all = forgetting_curve_df[
    forgetting_curve_df['learning_language'] == 'all'
].groupby(['lag_bin', 'lag_bin_order'])['avg_session_recall'].mean().reset_index()
fc_all = fc_all.sort_values('lag_bin_order')
fc_all['avg_session_recall_%'] = (fc_all['avg_session_recall'] * 100).round(2)
print(fc_all[['lag_bin', 'avg_session_recall_%']].to_string(index=False))
```

```python
# ============================================================
# H2 — STEP 1: PRIOR EXPOSURES — DISTRIBUTION & CORRELATION
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Load data ────────────────────────────────────────────────
traces_df = pd.read_csv("traces_cleaned.csv")

print("=" * 60)
print("H2 — STEP 1: PRIOR EXPOSURES vs RECALL")
print("=" * 60)

# ── 1. history_seen distribution ─────────────────────────────
print("\n── history_seen (prior exposures) distribution ──")
percentiles = [0, 10, 25, 50, 75, 90, 95, 99, 100]
for p in percentiles:
    print(f"  P{p:>3}  : {np.percentile(traces_df['history_seen'], p):.0f} exposures")

# ── 2. Cap at P99 to remove extreme outliers ─────────────────
exp_cap = np.percentile(traces_df['history_seen'], 99)
df_capped = traces_df[traces_df['history_seen'] <= exp_cap].copy()
print(f"\nRecords after P99 exposure cap ({exp_cap:.0f}): {len(df_capped):,}")

# ── 3. Spearman correlation ───────────────────────────────────
print("\n── Spearman Correlation: history_seen vs p_recall ──")
rho_full, p_full = stats.spearmanr(
    traces_df['history_seen'], traces_df['p_recall']
)
rho_capped, p_capped = stats.spearmanr(
    df_capped['history_seen'], df_capped['p_recall']
)
print(f"  Full dataset  : rho = {rho_full:.4f},  p-value = {p_full:.2e}")
print(f"  P99 capped    : rho = {rho_capped:.4f},  p-value = {p_capped:.2e}")

# ── 4. Direction check ────────────────────────────────────────
alpha = 0.05
print("\n── H2 Step 1 Initial Signal ──")
if p_capped < alpha and rho_capped > 0:
    print("  Result : SUPPORTS H2 — more exposures = higher recall")
elif p_capped < alpha and rho_capped < 0:
    print("  Result : CONTRADICTS H2 — more exposures = lower recall (unexpected)")
else:
    print("  Result : INSUFFICIENT EVIDENCE")

# ── 5. Mean recall by exposure decile ────────────────────────
print("\n── Mean Recall by Exposure Decile ──")
df_capped['exp_decile'] = pd.qcut(
    df_capped['history_seen'], q=10, duplicates='drop'
)
decile_recall = (
    df_capped.groupby('exp_decile', observed=True)['p_recall']
    .agg(['mean', 'count'])
    .reset_index()
)
decile_recall.columns = ['exp_decile', 'avg_recall', 'n_records']
decile_recall['avg_recall_%'] = (decile_recall['avg_recall'] * 100).round(2)
print(decile_recall[['exp_decile', 'avg_recall_%', 'n_records']].to_string(index=False))

# ── 6. First-time vs repeat exposure ─────────────────────────
print("\n── First-Time vs Repeat Exposure Recall ──")
first_time = traces_df[traces_df['history_seen'] == 1]['p_recall']
repeat     = traces_df[traces_df['history_seen'] > 1]['p_recall']
t_stat, t_p = stats.mannwhitneyu(repeat, first_time, alternative='greater')
print(f"  First-time exposure (n={len(first_time):,}) : {first_time.mean()*100:.2f}%")
print(f"  Repeat exposure    (n={len(repeat):,}) : {repeat.mean()*100:.2f}%")
print(f"  Mann-Whitney U p-value : {t_p:.4f}")
print(f"  Significant difference : {'Yes ✓' if t_p < alpha else 'No ✗'}")
```

```python
# ============================================================
# H2 — STEP 2: PRACTICE BINS + LAG CONTROLLED ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Load data ────────────────────────────────────────────────
traces_df          = pd.read_csv("traces_cleaned.csv")
forgetting_curve_df = pd.read_csv("forgetting_curve_cleaned.csv")
word_difficulty_df  = pd.read_csv("word_difficulty_cleaned.csv")

print("=" * 60)
print("H2 — STEP 2: PRACTICE BINS + CONTROLLED ANALYSIS")
print("=" * 60)

# ── 1. Create practice bins matching forgetting curve ────────
practice_bins   = [0, 2, 4, 9, 19, 99999]
practice_labels = [
    '1-2 exposures', '3-4 exposures', '5-9 exposures',
    '10-19 exposures', '20+ exposures'
]
practice_order = {label: i for i, label in enumerate(practice_labels)}

traces_df['practice_bin'] = pd.cut(
    traces_df['history_seen'],
    bins=practice_bins,
    labels=practice_labels,
    right=True
)

# ── 2. Recall by practice bin (raw) ─────────────────────────
print("\n── Recall by Practice Bin (Raw, No Lag Control) ──")
pb_raw = (
    traces_df.groupby('practice_bin', observed=True)['p_recall']
    .agg(['mean', 'std', 'count'])
    .reset_index()
)
pb_raw.columns = ['practice_bin', 'avg_recall', 'std', 'n_records']
pb_raw['avg_recall_%'] = (pb_raw['avg_recall'] * 100).round(2)
pb_raw['bin_order']    = pb_raw['practice_bin'].map(practice_order)
pb_raw = pb_raw.sort_values('bin_order')
print(pb_raw[['practice_bin', 'avg_recall_%', 'std', 'n_records']].to_string(index=False))

# ── 3. Kruskal-Wallis across practice bins ───────────────────
print("\n── Kruskal-Wallis Test Across Practice Bins ──")
groups = [
    traces_df[traces_df['practice_bin'] == pb]['p_recall'].dropna().values
    for pb in practice_labels
]
kw_stat, kw_p = stats.kruskal(*groups)
print(f"  H-statistic : {kw_stat:.2f}")
print(f"  p-value     : {kw_p:.2e}")
print(f"  Decision    : {'REJECT H0 — bins differ' if kw_p < 0.05 else 'FAIL TO REJECT H0'}")

# ── 4. Lag-controlled analysis ───────────────────────────────
# Hold lag constant by looking within the same lag bin
lag_bins   = [0, 1/24, 6/24, 1, 3, 7, 14, 28, 91, 9999]
lag_labels = [
    '<1 hour', '1-6 hours', '6-24 hours',
    '1-3 days', '3-7 days', '1-2 weeks',
    '2-4 weeks', '1-3 months', '3+ months'
]
traces_df['lag_bin'] = pd.cut(
    traces_df['lag_days'],
    bins=lag_bins,
    labels=lag_labels,
    right=True
)

print("\n── Recall by Practice Bin WITHIN each Lag Bin ──")
print("(Holding lag constant to isolate exposure effect)")
pivot = (
    traces_df.groupby(['lag_bin', 'practice_bin'], observed=True)['p_recall']
    .mean()
    .mul(100)
    .round(2)
    .unstack('practice_bin')
)
print(pivot.to_string())

# ── 5. Forgetting curve dataset — practice bin effect ────────
print("\n── Forgetting Curve Dataset: Recall by Practice Bin (All Languages) ──")
fc_all = forgetting_curve_df[
    forgetting_curve_df['learning_language'] == 'all'
].groupby(['practice_bin', 'practice_bin_order'])['avg_session_recall'].mean().reset_index()
fc_all = fc_all.sort_values('practice_bin_order')
fc_all['avg_session_recall_%'] = (fc_all['avg_session_recall'] * 100).round(2)
print(fc_all[['practice_bin', 'avg_session_recall_%']].to_string(index=False))

# ── 6. Word difficulty dataset — avg_prior_exposures ─────────
print("\n── Word Difficulty: Avg Prior Exposures by Recall Quartile ──")
word_difficulty_df['recall_quartile'] = pd.qcut(
    word_difficulty_df['item_recall_rate'],
    q=4,
    labels=['Q1 (hardest)', 'Q2', 'Q3', 'Q4 (easiest)']
)
exp_by_quartile = (
    word_difficulty_df.groupby('recall_quartile', observed=True)
    ['avg_prior_exposures']
    .agg(['mean', 'median', 'count'])
    .reset_index()
)
exp_by_quartile.columns = ['recall_quartile', 'avg_exposures', 'median_exposures', 'n_words']
print(exp_by_quartile.to_string(index=False))

print("\n── Key Question ──")
print("Do harder words (lower recall) need MORE exposures?")
print("If yes → high exposure count = proxy for difficulty, not mastery")
```

```python
# ============================================================
# H2 — STEP 3: VISUALISATION + FINAL CONCLUSION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import warnings
warnings.filterwarnings('ignore')

# ── Load data ────────────────────────────────────────────────
traces_df           = pd.read_csv("traces_cleaned.csv")
forgetting_curve_df = pd.read_csv("forgetting_curve_cleaned.csv")
word_difficulty_df  = pd.read_csv("word_difficulty_cleaned.csv")

# ── Recreate bins ────────────────────────────────────────────
practice_bins   = [0, 2, 4, 9, 19, 99999]
practice_labels = [
    '1-2', '3-4', '5-9', '10-19', '20+'
]
practice_labels_full = [
    '1-2 exposures', '3-4 exposures', '5-9 exposures',
    '10-19 exposures', '20+ exposures'
]

lag_bins   = [0, 1/24, 6/24, 1, 3, 7, 14, 28, 91, 9999]
lag_labels = [
    '<1 hour', '1-6 hours', '6-24 hours',
    '1-3 days', '3-7 days', '1-2 weeks',
    '2-4 weeks', '1-3 months', '3+ months'
]

traces_df['practice_bin'] = pd.cut(
    traces_df['history_seen'],
    bins=practice_bins,
    labels=practice_labels_full,
    right=True
)
traces_df['lag_bin'] = pd.cut(
    traces_df['lag_days'],
    bins=lag_bins,
    labels=lag_labels,
    right=True
)

# ── Build pivot for heatmap ──────────────────────────────────
pivot = (
    traces_df.groupby(['lag_bin', 'practice_bin'], observed=True)['p_recall']
    .mean()
    .mul(100)
    .round(2)
    .unstack('practice_bin')
)

# ── Word difficulty: exposures by recall quartile ────────────
word_difficulty_df['recall_quartile'] = pd.qcut(
    word_difficulty_df['item_recall_rate'],
    q=4,
    labels=['Q1\n(hardest)', 'Q2', 'Q3', 'Q4\n(easiest)']
)
exp_by_quartile = (
    word_difficulty_df.groupby('recall_quartile', observed=True)
    ['avg_prior_exposures']
    .mean()
    .reset_index()
)

# ── Forgetting curve practice bin data ───────────────────────
fc_all = forgetting_curve_df[
    forgetting_curve_df['learning_language'] == 'all'
].groupby(['practice_bin', 'practice_bin_order'])['avg_session_recall'].mean().reset_index()
fc_all = fc_all.sort_values('practice_bin_order')
fc_all['avg_session_recall_%'] = fc_all['avg_session_recall'] * 100

# ── Raw practice bin recall ───────────────────────────────────
pb_raw = (
    traces_df.groupby('practice_bin', observed=True)['p_recall']
    .mean()
    .mul(100)
    .reset_index()
)
pb_raw.columns = ['practice_bin', 'avg_recall_%']

# ════════════════════════════════════════════════════════════
# FIGURE: 3 panels
# ════════════════════════════════════════════════════════════
fig = plt.figure(figsize=(18, 6))
fig.suptitle(
    'H2 — Prior Exposures vs Word Recall Rate',
    fontsize=15, fontweight='bold', y=1.02
)

# ── PANEL 1: Raw recall by practice bin (bar) ────────────────
ax1 = fig.add_subplot(1, 3, 1)
colors_bar = ['#D5DBDB', '#AEB6BF', '#5D6D7E', '#2E86C1', '#C0392B']
bars = ax1.bar(
    practice_labels,
    pb_raw['avg_recall_%'],
    color=colors_bar,
    edgecolor='white',
    linewidth=0.8
)
for bar, val in zip(bars, pb_raw['avg_recall_%']):
    ax1.text(
        bar.get_x() + bar.get_width() / 2,
        val + 0.05,
        f'{val:.2f}%',
        ha='center', va='bottom',
        fontsize=8.5, fontweight='bold', color='#2C3E50'
    )
ax1.set_ylim(88.5, 90.5)
ax1.set_xlabel('Prior Exposures', fontsize=10)
ax1.set_ylabel('Average Recall Rate (%)', fontsize=10)
ax1.set_title('Raw Recall by\nExposure Count', fontsize=11)
ax1.grid(axis='y', linestyle='--', alpha=0.5)
ax1.spines[['top', 'right']].set_visible(False)

# Annotate peak
ax1.annotate(
    'Peak at\n10–19',
    xy=(3, 89.94), xytext=(3, 90.25),
    ha='center', fontsize=8, color='#2E86C1',
    arrowprops=dict(arrowstyle='->', color='#2E86C1', lw=1.2)
)
ax1.annotate(
    'Drops\nat 20+',
    xy=(4, 89.23), xytext=(3.6, 88.7),
    ha='center', fontsize=8, color='#C0392B',
    arrowprops=dict(arrowstyle='->', color='#C0392B', lw=1.2)
)

# ── PANEL 2: Heatmap — lag × practice bin ───────────────────
ax2 = fig.add_subplot(1, 3, 2)
import matplotlib.colors as mcolors

heatmap_data = pivot.values
im = ax2.imshow(
    heatmap_data,
    cmap='RdYlGn',
    aspect='auto',
    vmin=84, vmax=93
)
ax2.set_xticks(range(len(practice_labels)))
ax2.set_xticklabels(practice_labels, fontsize=8)
ax2.set_yticks(range(len(lag_labels)))
ax2.set_yticklabels(lag_labels, fontsize=8)
ax2.set_xlabel('Prior Exposures', fontsize=10)
ax2.set_ylabel('Time Since Last Practice', fontsize=10)
ax2.set_title('Recall Heatmap\n(Lag × Exposures)', fontsize=11)

for i in range(len(lag_labels)):
    for j in range(len(practice_labels)):
        val = heatmap_data[i, j]
        if not np.isnan(val):
            ax2.text(j, i, f'{val:.1f}', ha='center', va='center',
                     fontsize=7, color='black', fontweight='bold')

plt.colorbar(im, ax=ax2, label='Recall %', shrink=0.8)

# ── PANEL 3: Hard words need more exposures ──────────────────
ax3 = fig.add_subplot(1, 3, 3)
quartile_labels = ['Q1\n(hardest)', 'Q2', 'Q3', 'Q4\n(easiest)']
colors_q = ['#C0392B', '#E67E22', '#F1C40F', '#2ECC71']
bars3 = ax3.bar(
    quartile_labels,
    exp_by_quartile['avg_prior_exposures'],
    color=colors_q,
    edgecolor='white',
    linewidth=0.8
)
for bar, val in zip(bars3, exp_by_quartile['avg_prior_exposures']):
    ax3.text(
        bar.get_x() + bar.get_width() / 2,
        val + 0.15,
        f'{val:.1f}x',
        ha='center', va='bottom',
        fontsize=9, fontweight='bold', color='#2C3E50'
    )
ax3.set_xlabel('Word Recall Quartile', fontsize=10)
ax3.set_ylabel('Avg Prior Exposures', fontsize=10)
ax3.set_title('Hard Words Need More\nExposures (Survivorship)', fontsize=11)
ax3.grid(axis='y', linestyle='--', alpha=0.5)
ax3.spines[['top', 'right']].set_visible(False)
ax3.set_ylim(0, 16)

# Arrow showing direction
ax3.annotate(
    '',
    xy=(0, 13.2), xytext=(3, 7.3),
    arrowprops=dict(arrowstyle='->', color='#7F8C8D', lw=1.5)
)
ax3.text(1.5, 11, 'Harder words\nseen more often', fontsize=8,
         color='#7F8C8D', ha='center')

plt.tight_layout()
plt.savefig('H2_exposures_vs_recall.png', dpi=150, bbox_inches='tight')
plt.show()
print("Plot saved as H2_exposures_vs_recall.png")

# ── FINAL SUMMARY ────────────────────────────────────────────
print("\n" + "=" * 60)
print("H2 — FINAL CONCLUSION")
print("=" * 60)
print(f"""
Hypothesis : More prior exposures = higher recall
Result     : PARTIALLY SUPPORTED — with important nuance ⚠️

Key Numbers:
  • Recall range across bins : 89.2% → 89.9% (< 1 pp spread)
  • Peak recall              : 10–19 exposures (89.94%)
  • Drop at 20+              : 89.23% (below even 1–2 exposures)
  • Hardest words (Q1)       : 13.0 avg exposures
  • Easiest words (Q4)       : 7.1 avg exposures

Core Insight — Survivorship Bias:
  • High exposure count ≠ mastery
  • High exposure count = proxy for word difficulty
  • The algorithm already re-shows hard words more —
    but those words remain hard regardless

Interaction Effect:
  • Long lag + many exposures = better recall than
    long lag + few exposures
  • Exposure history only meaningfully protects recall
    when the review gap is very long (3+ months)

Business Implication:
  • Do NOT use raw exposure count as a mastery signal
  • Combine exposure count WITH recent recall accuracy
    to distinguish "seen often because hard" from
    "seen often and now mastered"
  • Words with 20+ exposures AND still low recall
    need a different intervention, not just more repetition
""")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ── H4 Step 1: Diminishing returns on exposures ──

df = word_difficulty_df.copy()

# Define exposure bins (unequal width — more granular at low end where change is fastest)
bins = [0, 2, 5, 9, 14, 19, 24, 29, 39, 59, 99, np.inf]
labels = ['1-2', '3-5', '6-9', '10-14', '15-19', '20-24', '25-29', '30-39', '40-59', '60-99', '100+']

df['exposure_bin'] = pd.cut(df['avg_prior_exposures'], bins=bins, labels=labels)

bin_summary = df.groupby('exposure_bin', observed=True).agg(
    avg_recall=('item_recall_rate', 'mean'),
    median_recall=('item_recall_rate', 'median'),
    word_count=('item_recall_rate', 'count'),
    avg_exposures=('avg_prior_exposures', 'mean')
).reset_index()

print("── Recall by Exposure Bin ──")
print(bin_summary.to_string(index=False))

# ── Marginal gain per bin ──
bin_summary['recall_gain'] = bin_summary['avg_recall'].diff().round(4)
print("\n── Marginal Recall Gain (vs previous bin) ──")
print(bin_summary[['exposure_bin', 'avg_recall', 'recall_gain', 'word_count']].to_string(index=False))

# ── Plot ──
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: recall curve
axes[0].plot(bin_summary['exposure_bin'], bin_summary['avg_recall'],
             marker='o', color='steelblue', linewidth=2.5, markersize=7)
axes[0].axhline(y=bin_summary['avg_recall'].max(), color='gray', linestyle='--', alpha=0.5, label='Peak recall')
axes[0].set_xlabel('Avg Prior Exposures (bin)')
axes[0].set_ylabel('Avg Recall Rate')
axes[0].set_title('Recall Rate by Exposure Count')
axes[0].tick_params(axis='x', rotation=45)
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Right: marginal gain bar chart
gain_data = bin_summary.dropna(subset=['recall_gain'])
colors = ['#2ecc71' if x >= 0 else '#e74c3c' for x in gain_data['recall_gain']]
axes[1].bar(gain_data['exposure_bin'], gain_data['recall_gain'], color=colors)
axes[1].axhline(y=0, color='black', linewidth=0.8)
axes[1].set_xlabel('Exposure Bin')
axes[1].set_ylabel('Marginal Recall Gain')
axes[1].set_title('Marginal Gain per Additional Exposure Bracket')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ── H4 Step 2: Scatter + polynomial fit on raw data ──

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

df_filtered = df[df['avg_prior_exposures'] <= 100].copy()  # remove extreme outliers

x = df_filtered['avg_prior_exposures'].values
y = df_filtered['item_recall_rate'].values

# Fit degree-2 polynomial (captures peak + fall)
coeffs = np.polyfit(x, y, 2)
p2 = np.poly1d(coeffs)

# Find peak of the curve
peak_x = -coeffs[1] / (2 * coeffs[0])
peak_y = p2(peak_x)

# Overall linear correlation
r, pval = pearsonr(x, y)
print(f"Linear correlation (exposures vs recall, capped at 100): r = {r:.3f}, p = {pval:.4f}")
print(f"\nQuadratic fit peak: {peak_x:.1f} exposures → predicted recall = {peak_y:.3f}")
print(f"Quadratic coefficients: a={coeffs[0]:.6f}, b={coeffs[1]:.5f}, c={coeffs[2]:.4f}")

# Plot
fig, ax = plt.subplots(figsize=(10, 5))

# Scatter (sample for readability)
sample = df_filtered.sample(min(3000, len(df_filtered)), random_state=42)
ax.scatter(sample['avg_prior_exposures'], sample['item_recall_rate'],
           alpha=0.2, color='steelblue', s=12, label='Words (sampled)')

# Polynomial curve
x_line = np.linspace(0, 100, 300)
ax.plot(x_line, p2(x_line), 'r-', linewidth=2.5, label=f'Quadratic fit (peak ≈ {peak_x:.0f} exposures)')

# Peak marker
ax.axvline(x=peak_x, color='orange', linestyle='--', linewidth=1.5, label=f'Peak at ~{peak_x:.0f} exposures')
ax.axhline(y=peak_y, color='orange', linestyle=':', linewidth=1)

ax.set_xlabel('Avg Prior Exposures')
ax.set_ylabel('Item Recall Rate')
ax.set_title('H4: Recall vs Exposure Count — Quadratic Fit')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# Ensure practice_time is datetime
traces_df['practice_time'] = pd.to_datetime(traces_df['practice_time'])

# --- Per-user consistency metrics ---
user_stats = traces_df.groupby('user_id').agg(
    mean_lag       = ('lag_days', 'mean'),
    std_lag        = ('lag_days', 'std'),
    total_sessions = ('p_recall', 'count'),
    mean_recall    = ('p_recall', 'mean')
).reset_index()

# Coefficient of variation: std/mean (lower = more consistent relative to their own average)
user_stats['cv_lag'] = user_stats['std_lag'] / user_stats['mean_lag']

# Drop users with only 1 session (can't compute std) or NaN cv
user_stats = user_stats.dropna(subset=['cv_lag'])

# Filter: at least 5 sessions (meaningful consistency signal)
user_stats = user_stats[user_stats['total_sessions'] >= 5]

print(f"Users remaining after filter: {len(user_stats)}")
print(f"\nSummary stats:")
print(user_stats[['mean_lag', 'std_lag', 'cv_lag', 'total_sessions', 'mean_recall']].describe().round(3))
```

```python
import matplotlib.pyplot as plt
import scipy.stats as stats

# --- Bucket users into 3 consistency tiers based on cv_lag percentiles ---
user_stats['consistency_tier'] = pd.qcut(
    user_stats['cv_lag'],
    q=3,
    labels=['High Consistency\n(low cv)', 'Medium Consistency', 'Low Consistency\n(high cv)']
)

# --- Summary by tier ---
tier_summary = user_stats.groupby('consistency_tier', observed=True).agg(
    n_users      = ('user_id', 'count'),
    mean_recall  = ('mean_recall', 'mean'),
    median_recall= ('mean_recall', 'median'),
    mean_lag     = ('mean_lag', 'mean'),
    mean_cv      = ('cv_lag', 'mean')
).round(4)

print("=== Recall by Consistency Tier ===")
print(tier_summary)

# --- Kruskal-Wallis test (non-parametric, since recall is skewed) ---
groups = [
    user_stats[user_stats['consistency_tier'] == tier]['mean_recall'].values
    for tier in user_stats['consistency_tier'].cat.categories
]
stat, p_value = stats.kruskal(*groups)
print(f"\nKruskal-Wallis H={stat:.2f}, p={p_value:.4f}")

# --- Bar chart ---
fig, ax = plt.subplots(figsize=(8, 5))
tiers = tier_summary.index.tolist()
recalls = tier_summary['mean_recall'].values

bars = ax.bar(tiers, recalls, color=['#2ecc71', '#f39c12', '#e74c3c'], edgecolor='black', width=0.5)
ax.set_ylim(0.85, 0.95)
ax.set_ylabel('Mean Recall Rate (p_recall)', fontsize=12)
ax.set_title('H7: Practice Consistency vs Recall Rate\n(cv_lag terciles)', fontsize=13)

for bar, val in zip(bars, recalls):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.001,
            f'{val:.4f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# --- Split users into frequency buckets first, THEN check consistency within each ---

# Frequency bucket based on mean_lag (how often they practice on average)
user_stats['freq_bucket'] = pd.qcut(
    user_stats['mean_lag'],
    q=3,
    labels=['Frequent\n(<3.4 days avg)', 'Moderate\n(3.4–8.4 days)', 'Infrequent\n(8.4+ days avg)']
)

# Within each frequency bucket, split by consistency (cv_lag median split)
user_stats['consistent'] = user_stats.groupby('freq_bucket', observed=True)['cv_lag'].transform(
    lambda x: pd.qcut(x, q=2, labels=['Consistent', 'Inconsistent'])
)

# Summary: recall by frequency × consistency
cross_summary = user_stats.groupby(['freq_bucket', 'consistent'], observed=True).agg(
    n_users     = ('user_id', 'count'),
    mean_recall = ('mean_recall', 'mean'),
    mean_cv     = ('cv_lag', 'mean'),
    mean_lag    = ('mean_lag', 'mean')
).round(4)

print("=== Recall by Frequency × Consistency ===")
print(cross_summary)

# Within each frequency bucket: does consistency matter?
print("\n=== Does consistency add anything within same frequency? ===")
for bucket in user_stats['freq_bucket'].cat.categories:
    subset = user_stats[user_stats['freq_bucket'] == bucket]
    g1 = subset[subset['consistent'] == 'Consistent']['mean_recall']
    g2 = subset[subset['consistent'] == 'Inconsistent']['mean_recall']
    stat, p = stats.mannwhitneyu(g1, g2, alternative='two-sided')
    diff = g1.mean() - g2.mean()
    print(f"{bucket}: Consistent={g1.mean():.4f} vs Inconsistent={g2.mean():.4f} | diff={diff:+.4f} | p={p:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Use 'all languages' aggregate for clean overall threshold ---
all_lang = forgetting_curve_df[
    forgetting_curve_df['learning_language'] == 'all'
].groupby(['lag_bin', 'lag_bin_order']).agg(
    mean_recall  = ('item_recall_rate', 'mean'),
    total_traces = ('n_traces', 'sum')
).reset_index().sort_values('lag_bin_order')

# --- Calculate drop from previous bin (marginal decay per step) ---
all_lang['recall_drop'] = all_lang['mean_recall'].diff()  # negative = decay
all_lang['pct_drop_from_start'] = (
    (all_lang['mean_recall'] - all_lang['mean_recall'].iloc[0]) * 100
).round(3)

print("=== Recall & Decay Rate Across Lag Bins ===")
print(all_lang[['lag_bin', 'mean_recall', 'recall_drop',
                 'pct_drop_from_start', 'total_traces']].to_string(index=False))

# --- Plot 1: Recall curve with decay annotations ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 8))

lag_labels = all_lang['lag_bin'].tolist()
x = range(len(lag_labels))

# Top: recall curve
ax1.plot(x, all_lang['mean_recall'], marker='o', linewidth=2.5,
         color='#2ecc71', markersize=8)
ax1.fill_between(x, all_lang['mean_recall'], alpha=0.15, color='#2ecc71')
ax1.set_xticks(x)
ax1.set_xticklabels(lag_labels, rotation=30, ha='right', fontsize=9)
ax1.set_ylabel('Mean Recall Rate', fontsize=11)
ax1.set_title('H11: Recall Decay Curve — Where Is the Cliff?', fontsize=13)
ax1.set_ylim(0.86, 0.94)
ax1.grid(axis='y', alpha=0.4)
for i, (xi, yi) in enumerate(zip(x, all_lang['mean_recall'])):
    ax1.text(xi, yi + 0.001, f'{yi:.4f}', ha='center', fontsize=8, fontweight='bold')

# Bottom: marginal drop per bin (bar chart)
drops = all_lang['recall_drop'].fillna(0).values
colors = ['#e74c3c' if d < -0.005 else '#f39c12' if d < 0 else '#2ecc71' for d in drops]
ax2.bar(x, drops * 100, color=colors, edgecolor='black')  # convert to pp
ax2.set_xticks(x)
ax2.set_xticklabels(lag_labels, rotation=30, ha='right', fontsize=9)
ax2.set_ylabel('Marginal Drop (percentage points)', fontsize=11)
ax2.set_title('Step-by-Step Decay: How Much Recall Is Lost Per Lag Window?', fontsize=12)
ax2.axhline(0, color='black', linewidth=0.8)
ax2.grid(axis='y', alpha=0.4)
for i, (xi, d) in enumerate(zip(x, drops)):
    if d != 0:
        ax2.text(xi, d * 100 - 0.05, f'{d*100:.2f}pp',
                 ha='center', va='top', fontsize=8, fontweight='bold', color='white')

plt.tight_layout()
plt.show()
```

```python
# --- Identify the steepest decay windows ---
print("=== Decay Acceleration Analysis ===")
print("Cumulative recall loss from <1 hour baseline:\n")

lag_order = ['<1 hour','1-6 hours','6-24 hours','1-3 days',
             '3-7 days','1-2 weeks','2-4 weeks','1-3 months','3+ months']

# Midpoint days for each bin (for threshold estimation)
lag_midpoints = {
    '<1 hour': 0.02, '1-6 hours': 0.15, '6-24 hours': 0.5,
    '1-3 days': 2, '3-7 days': 5, '1-2 weeks': 10,
    '2-4 weeks': 21, '1-3 months': 45, '3+ months': 120
}
all_lang['midpoint_days'] = all_lang['lag_bin'].map(lag_midpoints)

# Acceptable recall threshold = 90% (business decision line)
THRESHOLD = 0.90
all_lang['below_threshold'] = all_lang['mean_recall'] < THRESHOLD

print(all_lang[['lag_bin', 'mean_recall', 'pct_drop_from_start', 'below_threshold']].to_string(index=False))
first_breach = all_lang[all_lang['below_threshold']]['lag_bin'].iloc[0]
print(f"\n⚠️  Recall drops below {THRESHOLD*100}% at: {first_breach}")

# --- Language-specific thresholds ---
print("\n=== Language-Specific: When Does Recall Drop Below 90%? ===")
lang_lag_full = forgetting_curve_df[
    forgetting_curve_df['learning_language'] != 'all'
].groupby(['learning_language', 'language_name', 'lag_bin', 'lag_bin_order']).agg(
    mean_recall = ('item_recall_rate', 'mean')
).reset_index().sort_values(['learning_language', 'lag_bin_order'])

lang_thresholds = []
for lang in lang_lag_full['language_name'].unique():
    subset = lang_lag_full[lang_lag_full['language_name'] == lang].sort_values('lag_bin_order')
    subset['midpoint_days'] = subset['lag_bin'].map(lag_midpoints)

    # Starting recall
    start_recall = subset.iloc[0]['mean_recall']

    # First bin below 90%
    below = subset[subset['mean_recall'] < THRESHOLD]
    breach_bin = below.iloc[0]['lag_bin'] if len(below) > 0 else 'Never drops below 90%'
    breach_days = below.iloc[0]['midpoint_days'] if len(below) > 0 else None

    # Total decay
    end_recall = subset.iloc[-1]['mean_recall']
    total_decay = (start_recall - end_recall) * 100

    lang_thresholds.append({
        'Language': lang,
        'Start Recall': round(start_recall, 4),
        'End Recall': round(end_recall, 4),
        'Total Decay (pp)': round(total_decay, 2),
        'Drops Below 90% At': breach_bin,
        'Approx Days': breach_days
    })

thresh_df = pd.DataFrame(lang_thresholds).sort_values('Approx Days')
print(thresh_df.to_string(index=False))

# --- Final chart: language decay curves with 90% threshold line ---
fig, ax = plt.subplots(figsize=(12, 6))

colors = {'German': '#e74c3c', 'French': '#e67e22', 'Spanish': '#2ecc71',
          'Portuguese': '#3498db', 'Italian': '#9b59b6', 'English': '#1abc9c'}

for lang in lang_lag_full['language_name'].unique():
    subset = lang_lag_full[lang_lag_full['language_name'] == lang].sort_values('lag_bin_order')
    ax.plot(subset['lag_bin_order'], subset['mean_recall'],
            marker='o', linewidth=2.5, label=lang, color=colors.get(lang, 'gray'))

# 90% threshold line
ax.axhline(y=THRESHOLD, color='red', linestyle='--', linewidth=2, label='90% threshold')
ax.fill_between(range(9), THRESHOLD, 0.84, alpha=0.08, color='red')

ax.set_xticks(range(len(lag_order)))
ax.set_xticklabels(lag_order, rotation=35, ha='right', fontsize=9)
ax.set_ylabel('Mean Recall Rate', fontsize=12)
ax.set_title('H11: Scheduling Sweet Spot — When Does Each Language Cross 90%?', fontsize=13)
ax.legend(fontsize=10)
ax.set_ylim(0.84, 0.94)
ax.grid(axis='y', alpha=0.4)

plt.tight_layout()
plt.show()
```
